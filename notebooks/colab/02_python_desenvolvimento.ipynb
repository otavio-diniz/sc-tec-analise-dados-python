# ✅ CHECKPOINT 1 — Comparações e decisões

## 1.1 Operadores de comparação

Operadores de comparação servem para **comparar dois valores**.

O resultado de uma comparação é sempre um valor booleano:

- `True` → verdadeiro;
- `False` → falso.

| Operador | Nome / leitura | Pergunta que ele faz |
|---|---|---|
| `==` | igualdade / **igual a** | os dois valores são iguais? |
| `!=` | diferença / **diferente de** | os dois valores são diferentes? |
| `>` | **maior que** | o valor da esquerda é maior? |
| `<` | **menor que** | o valor da esquerda é menor? |
| `>=` | **maior ou igual a** | é maior ou igual? |
| `<=` | **menor ou igual a** | é menor ou igual? |

### Atenção a `=` e `==`

```python
idade = 28
```

`=` faz **atribuição**: a variável `idade` recebe `28`.

```python
idade == 28
```

`==` faz **comparação de igualdade**: o Python verifica se `idade` contém um valor igual a `28`.


````python
idade = 28

print("idade == 28  →", idade == 28)   # igualdade
print("idade != 30  →", idade != 30)   # diferente de
print("idade > 18   →", idade > 18)    # maior que
print("idade < 18   →", idade < 18)    # menor que
print("idade >= 28  →", idade >= 28)   # maior ou igual
print("idade <= 27  →", idade <= 27)   # menor ou igual
````

### Como ler o resultado

Se executarmos:

```python
idade = 28
print(idade >= 18)
```

a pergunta feita ao Python é:

> “28 é maior ou igual a 18?”

Como a resposta é verdadeira, o resultado é:

```text
True
```

Comparações são fundamentais porque estruturas como `if` usam exatamente esse tipo de resultado para decidir qual caminho executar.


## 1.2 Operadores lógicos

Operadores lógicos permitem **combinar ou inverter condições**.

| Operador | Nome / leitura | O que faz |
|---|---|---|
| `and` | **e** | resulta em `True` somente quando **todas** as condições são verdadeiras |
| `or` | **ou** | resulta em `True` quando **pelo menos uma** condição é verdadeira |
| `not` | **não / negação** | inverte o valor lógico: `True` vira `False` e `False` vira `True` |

Exemplo:

```python
idade >= 18 and tem_documento
```

Leia como:

> “A idade é pelo menos 18 **E** a pessoa tem documento?”


````python
idade = 28
tem_documento = True

print("Maior de idade E tem documento:", idade >= 18 and tem_documento)
print("Menor de idade OU tem documento:", idade < 18 or tem_documento)
print("NÃO tem documento:", not tem_documento)
````

## 1.3 `if`, `elif` e `else`

Essas palavras formam uma **estrutura condicional**: o programa verifica condições e escolhe qual bloco executar.

| Palavra | Leitura | Função |
|---|---|---|
| `if` | **se** | testa a primeira condição |
| `elif` | **senão se** | testa outra condição quando as anteriores foram falsas |
| `else` | **senão** | executa quando nenhuma condição anterior foi verdadeira |

### Estrutura básica

```python
if condição_1:
    # executa se condição_1 for True
elif condição_2:
    # executa se condição_1 for False e condição_2 for True
else:
    # executa se nenhuma condição anterior for True
```

### E o `:`?

Os dois-pontos `:` indicam que começará um **bloco de código** associado àquela condição.

### E o recuo dos espaços?

As linhas que pertencem ao bloco ficam **indentadas**, normalmente com quatro espaços. Em Python, essa indentação faz parte da estrutura do código.

O Python testa as condições **de cima para baixo** e, ao encontrar a primeira condição verdadeira, executa aquele bloco e não testa os blocos seguintes daquela mesma cadeia.


````python
valor = 180

if valor >= 250:
    classificacao = "ALTO"
elif valor >= 100:
    classificacao = "MÉDIO"
else:
    classificacao = "BAIXO"

print("Valor:", valor)
print("Classificação:", classificacao)
````

### Lendo o exemplo passo a passo

Com `valor = 180`:

1. `valor >= 250` → `False`.
2. O Python passa para `elif valor >= 100`.
3. `180 >= 100` → `True`.
4. A variável `classificacao` recebe `"MÉDIO"`.
5. O `else` não é executado.

Experimente trocar `180` por `300` e depois por `50` para observar os três caminhos.


### 💡 Dica de Ouro — a regra de negócio vem antes do operador

Em aula surgiu uma diferença pequena na escrita, mas grande no resultado:

> “desconto **a partir de 200**”

“A partir de 200” inclui o próprio `200`. Portanto:

```python
valor >= 200
```

é coerente; `valor > 200` excluiria exatamente `200`.

Em programação e análise de dados, **interpretar o enunciado faz parte da solução**. Um código pode executar sem erro e ainda responder à regra errada.


## 1.4 Expressão condicional em uma linha — “operador ternário”

Na aula de 22/09 apareceu a ideia de **encurtar uma decisão simples** usando uma expressão condicional:

```python
desconto = valor * 0.10 if valor >= 200 else 0
```

Leia em três blocos:

1. `valor * 0.10` → resultado quando a condição é verdadeira;
2. `if valor >= 200` → condição;
3. `else 0` → resultado quando a condição é falsa.

A forma equivalente, mais longa, é:

```python
if valor >= 200:
    desconto = valor * 0.10
else:
    desconto = 0
```

A expressão em uma linha é útil quando a regra é **curta e fácil de ler**. Se a decisão exigir várias etapas, validações ou efeitos diferentes, prefira `if`/`elif`/`else` em blocos.

> 💡 **Regra de legibilidade:** código menor não é automaticamente código melhor. Use a forma que deixa a regra mais clara.

# ✅ CHECKPOINT 2 — Coleções

Uma **coleção** permite guardar vários valores dentro de uma mesma estrutura.

Neste checkpoint vamos trabalhar com três estruturas muito comuns:

- **lista** → `[]`;
- **tupla** → `()`;
- **dicionário** → `{chave: valor}`.

Elas podem parecer semelhantes porque todas guardam informações, mas organizam e acessam esses dados de maneiras diferentes.


## 2.1 Lista `[]`

Uma **lista** é uma coleção **ordenada e mutável**.

- **ordenada:** cada item possui uma posição;
- **mutável:** podemos adicionar, remover ou alterar itens depois que a lista foi criada.

A forma literal mais comum de criar uma lista usa **colchetes `[]`**:

```python
vendas = [120.0, 85.5, 230.0]
```

As vírgulas separam os elementos.

> Você também pode encontrar listas criadas de outras formas, como `list(...)`, mas `[]` é a forma visual mais comum e importante neste momento.

### O que significa `.append()`?

`append` é um **método de lista** usado para adicionar **um novo elemento ao final** da lista.

```python
vendas.append(310.0)
```

Leia assim:

> “Na lista `vendas`, adicione `310.0` ao final.”

O ponto `.` conecta o objeto `vendas` a uma operação que essa lista sabe realizar: `append()`.
### `len()` e `sum()` — contar itens × somar valores

- `len(vendas)` informa **quantos elementos** existem na lista.
- `sum(vendas)` soma os valores numéricos da lista.

```python
vendas = [120.0, 85.5, 230.0]
print(len(vendas))  # 3
print(sum(vendas))  # 435.5
```

`len()` responde “quantos itens?”. `sum()` responde “quanto dá a soma?”.


````python
vendas = [120.0, 85.5, 230.0]

print("Lista original:", vendas)

vendas.append(310.0)

print("Depois do append:", vendas)
print("quantidade de vendas:", len(vendas))
print("soma das vendas:", sum(vendas))
````

O resultado final é:

```text
[120.0, 85.5, 230.0, 310.0]
```

O novo valor aparece no **fim** porque esse é o comportamento de `append()`.


## 2.2 Indexação: por que a contagem começa em zero?

Em Python, posições de sequências como listas começam no índice **0**, e não no índice 1.

Para:

```python
nomes = ["Ana", "Bruno", "Carla"]
```

podemos visualizar assim:

| Índice | Valor |
|---:|---|
| `0` | `"Ana"` |
| `1` | `"Bruno"` |
| `2` | `"Carla"` |

Portanto:

- `nomes[0]` → primeiro item → `"Ana"`;
- `nomes[1]` → segundo item → `"Bruno"`;
- `nomes[2]` → terceiro item → `"Carla"`.

### Por que zero?

Essa convenção é chamada de **indexação baseada em zero**. Para quem está começando, a regra prática é:

> **primeiro item = índice 0**.

Assim, uma lista com três itens possui índices `0`, `1` e `2`.


````python
nomes = ["Ana", "Bruno", "Carla"]

print("Índice 0:", nomes[0])
print("Índice 1:", nomes[1])
print("Índice 2:", nomes[2])
````

### Índices negativos

Python também permite contar **de trás para frente**:

| Índice negativo | Valor |
|---:|---|
| `-1` | último item → `"Carla"` |
| `-2` | penúltimo item → `"Bruno"` |
| `-3` | antepenúltimo item → `"Ana"` |

Por isso:

```python
nomes[-1]
```

retorna:

```text
Carla
```

O `-1` não significa “posição menos um”. Nesse contexto, ele é uma forma prática de pedir **o último elemento**.


````python
nomes = ["Ana", "Bruno", "Carla"]

print("Último item com [-1]:", nomes[-1])
print("Penúltimo item com [-2]:", nomes[-2])
````

### Fatiamento: `nomes[0:2]`

Quando usamos:

```python
nomes[0:2]
```

estamos fazendo um **slice** (fatiamento).

Leia como:

> “Pegue os itens começando no índice `0` e pare **antes** do índice `2`.”

Portanto entram:

- índice `0` → `"Ana"`;
- índice `1` → `"Bruno"`.

O índice `2` (`"Carla"`) **não entra**, porque o limite final do slice é exclusivo.

Resultado:

```python
['Ana', 'Bruno']
```

### Por que o resultado aparece com colchetes e aspas?

Porque o fatiamento devolve **uma nova lista**.

- `[` e `]` mostram que o resultado é uma lista;
- `'Ana'` e `'Bruno'` aparecem com aspas porque cada elemento é uma string.

Não recebemos apenas um texto: recebemos uma coleção contendo dois textos.


````python
nomes = ["Ana", "Bruno", "Carla"]

parte = nomes[0:2]

print("Resultado do slice:", parte)
print("Tipo do resultado:", type(parte))
print("Quantidade de itens:", len(parte))
````

## 2.3 Tupla `()`

Uma **tupla** também é uma coleção ordenada, mas sua principal característica é ser **imutável**.

Isso significa que, depois de criada, não alteramos seus elementos da mesma forma que fazemos com uma lista.

A forma visual mais comum é:

```python
coordenada = (-22.90, -43.20)
```

Aqui temos dois valores na mesma estrutura.

### Quando uma tupla pode fazer sentido?

Quando queremos representar um conjunto de valores que forma uma unidade e que, naquele contexto, não deveria ser modificado diretamente.

Exemplos:

- latitude e longitude;
- uma data representada por partes;
- valores devolvidos juntos por uma função.

### Parênteses sempre significam tupla?

Não.

Você já viu `()` em chamadas de função:

```python
print("Olá")
```

Por isso, o **contexto** continua sendo importante.

Em uma tupla, os valores são separados por vírgulas. Os parênteses tornam essa estrutura mais clara visualmente.


````python
coordenada = (-22.90, -43.20)

print("Tupla completa:", coordenada)
print("Primeiro valor:", coordenada[0])
print("Segundo valor:", coordenada[1])
print("Tipo:", type(coordenada))

latitude, longitude = coordenada

print("Latitude:", latitude)
print("Longitude:", longitude)
````

No trecho:

```python
latitude, longitude = coordenada
```

acontece um **desempacotamento**:

- o primeiro valor da tupla vai para `latitude`;
- o segundo valor vai para `longitude`.


## 2.4 Dicionário `{chave: valor}`

Um **dicionário** organiza informações em pares:

```text
chave → valor
```

Exemplo:

```python
produto = {
    "nome": "Teclado",
    "preco": 249.90,
    "estoque": 7
}
```

Nesse dicionário:

- `"nome"` é uma chave e `"Teclado"` é seu valor;
- `"preco"` é uma chave e `249.90` é seu valor;
- `"estoque"` é uma chave e `7` é seu valor.

### Como reconhecer a estrutura?

- `{` e `}` delimitam o dicionário;
- `:` separa cada **chave** do seu **valor**;
- `,` separa um par do próximo.

### Por que o código foi escrito em várias linhas e com espaços?

Essa formatação é feita principalmente para **legibilidade**.

Dentro das chaves, este código:

```python
produto = {
    "nome": "Teclado",
    "preco": 249.90,
    "estoque": 7
}
```

representa a mesma estrutura que:

```python
produto = {"nome": "Teclado", "preco": 249.90, "estoque": 7}
```

A versão em várias linhas fica mais fácil de ler quando há vários campos.

Os quatro espaços antes de cada linha interna são uma escolha de organização visual nesse caso. Diferentemente dos blocos de `if`, aqui eles **não criam um bloco condicional**.

### Como acessar um valor?

Usamos a chave entre colchetes:

```python
produto["nome"]
```

Leia como:

> “No dicionário `produto`, acesse o valor associado à chave `"nome"`.”


````python
produto = {
    "nome": "Teclado",
    "preco": 249.90,
    "estoque": 7
}

print("Dicionário completo:", produto)
print("Nome:", produto["nome"])
print("Preço:", produto["preco"])
print("Estoque:", produto["estoque"])
````

## 2.5 Estruturas podem ser combinadas

Na prática, dados reais raramente aparecem isolados.

Podemos combinar estruturas para representar informações mais completas.

Imagine uma pequena lista de produtos. Cada produto possui várias características:

- nome;
- preço;
- estoque.

Uma forma natural de modelar isso é:

- usar **uma lista** para guardar vários produtos;
- usar **um dicionário** para representar cada produto.

Visualmente:

```text
lista
├── dicionário do produto 1
├── dicionário do produto 2
└── dicionário do produto 3
```

Assim, cada estrutura cumpre uma função:

- `[]` organiza a sequência de produtos;
- `{}` organiza os campos de cada produto;
- `["nome"]`, `["preco"]` e `["estoque"]` acessam informações específicas de um produto.


````python
produtos = [
    {
        "nome": "Mouse",
        "preco": 89.90,
        "estoque": 14
    },
    {
        "nome": "Teclado",
        "preco": 249.90,
        "estoque": 7
    },
    {
        "nome": "Webcam",
        "preco": 179.90,
        "estoque": 5
    }
]

print("Todos os produtos:")
print(produtos)

print()
print("Segundo produto completo:")
print(produtos[1])

print()
print("Nome do segundo produto:")
print(produtos[1]["nome"])

print()
print("Estoque do terceiro produto:")
print(produtos[2]["estoque"])
````

### Como ler `produtos[1]["nome"]`

Faça a leitura em duas etapas:

1. `produtos[1]` → pegue o item de índice `1`, isto é, o **segundo produto**;
2. `["nome"]` → nesse dicionário, pegue o valor associado à chave `"nome"`.

Resultado:

```text
Teclado
```

Esse exemplo mostra por que conhecer bem **índices** e **chaves** é importante antes de combinar estruturas.


## 2.6 Estruturas compostas e aninhadas

A aula de 02/10 avançou do dicionário isolado para **estruturas que combinam várias camadas**.

### Lista de dicionários

```python
produtos = [
    {"nome": "Mouse", "preco": 89.90, "estoque": 14},
    {"nome": "Teclado", "preco": 249.90, "estoque": 7},
    {"nome": "Webcam", "preco": 179.90, "estoque": 3},
]
```

Leia a estrutura de fora para dentro:

- `[...]` → uma lista que reúne vários registros;
- cada `{...}` → um produto descrito como dicionário;
- `"nome"`, `"preco"` e `"estoque"` → chaves;
- os valores após `:` → dados daquele produto.

Acesso encadeado:

```python
produtos[1]["preco"]
```

1. `produtos[1]` seleciona o **segundo dicionário** da lista;
2. `["preco"]` consulta a chave `"preco"` dentro dele.

### Dicionário contendo outro dicionário e uma lista

```python
pedido = {
    "id": 1001,
    "cliente": {
        "nome": "Ana Silva",
        "estado": "RJ",
    },
    "itens": [
        {"produto": "Mouse", "quantidade": 2, "preco": 89.90},
        {"produto": "Hub USB-C", "quantidade": 1, "preco": 159.90},
    ],
}
```

Agora podemos ler:

```python
pedido["cliente"]["nome"]
pedido["itens"][0]["produto"]
```

Isso exige seguir o caminho na mesma ordem em que os contêineres aparecem.

# ✅ CHECKPOINT 3 — Repetição

Até aqui, executamos instruções uma vez ou escolhemos caminhos com `if`. Agora entra outra necessidade comum: **repetir uma ação sem copiar a mesma linha várias vezes**.

Uma repetição também é chamada de **loop** ou **laço de repetição**. Cada passagem pelo bloco repetido recebe o nome de **iteração**.

Neste checkpoint veremos:

1. `for` — percorrer itens;
2. `range()` — gerar uma sequência de números inteiros;
3. `enumerate()` — obter posição e valor ao mesmo tempo;
4. `while` — repetir enquanto uma condição for verdadeira;
5. `break` e `continue` — controlar o andamento de um loop.


## 3.1 `for` — percorrer itens de uma coleção

Use `for` quando queremos **pegar os elementos de uma coleção, um por vez**, e executar uma ação para cada elemento.

Estrutura básica:

```python
for item in colecao:
    # código repetido
```

Leia como:

> “Para cada `item` que existir em `colecao`, execute o bloco indentado.”

No exemplo:

```python
vendas = [120, 85, 230]

for venda in vendas:
    print(venda)
```

- `vendas` é a lista completa;
- `venda` é uma **variável temporária** que recebe um item por vez;
- `in` pode ser lido aqui como **“em”** ou **“dentro de”**;
- `:` indica o início do bloco do `for`;
- a indentação mostra quais linhas serão repetidas.

A variável `venda` não precisa existir antes. Ela é criada/atualizada pelo próprio `for` a cada iteração.


````python
vendas = [120, 85, 230]

for venda in vendas:
    print("Valor da venda nesta iteração:", venda)
````

### O que acontece passo a passo?

Com `vendas = [120, 85, 230]`:

1. primeira iteração → `venda` recebe `120`;
2. segunda iteração → `venda` recebe `85`;
3. terceira iteração → `venda` recebe `230`;
4. não existem mais itens → o `for` termina.

Por isso aparecem três linhas de saída.


## 3.2 `range()` — gerar uma sequência de inteiros

A forma completa é:

```python
range(início, fim, passo)
```

- **início** → primeiro valor;
- **fim** → limite que **não entra**;
- **passo** → quanto avança ou retrocede a cada iteração.

### Forma curta: `range(5)`

Quando há apenas um número, o Python assume início `0` e passo `1`:

```text
0, 1, 2, 3, 4
```

O `5` é o limite final e não é incluído.


````python
for numero in range(5):
    print(numero)
````

### Início, fim e passo

```python
range(2, 6)
```

gera `2, 3, 4, 5`.

```python
range(1, 10, 2)
```

gera `1, 3, 5, 7, 9`.

O passo pode ser negativo:

```python
range(5, 0, -1)
```

gera `5, 4, 3, 2, 1`.

> 💡 **Dica de Ouro:** no `range(início, fim, passo)`, o **fim é exclusivo**. Pergunte sempre: onde começa, antes de qual valor para e de quanto em quanto anda?


````python
print("range(2, 6):")
for numero in range(2, 6):
    print(numero)

print("\nrange(1, 10, 2):")
for numero in range(1, 10, 2):
    print(numero)

print("\nrange(5, 0, -1):")
for numero in range(5, 0, -1):
    print(numero)
````

## 3.3 Contador × acumulador

Os dois podem começar em zero, mas respondem perguntas diferentes.

### Contador — “quantos?”

```python
aprovados = 0
for nota in [7.5, 5.0, 8.0, 9.0]:
    if nota >= 7:
        aprovados += 1
```

O contador normalmente soma **1** quando o evento acontece.

### Acumulador — “quanto ao todo?”

```python
total = 0
for venda in [100, 250, 80]:
    total += venda
```

O acumulador soma os **valores** encontrados.

> 💡 **Dica de Ouro da aula:** contador responde **quantos?**; acumulador responde **quanto ao todo?**


````python
notas = [7.5, 5.0, 8.0, 9.0]
aprovados = 0
for nota in notas:
    if nota >= 7:
        aprovados += 1

vendas = [100, 250, 80]
total = 0
for venda in vendas:
    total += venda

print("Quantidade de aprovados:", aprovados)
print("Total de vendas:", total)
````

## 3.4 `enumerate()` — posição e valor ao mesmo tempo **(complemento de apoio)**

Às vezes não basta saber o valor; também queremos saber **em qual posição ele está**.

`enumerate()` recebe uma coleção e, durante o `for`, fornece dois dados:

1. o **índice**;
2. o **valor**.

Exemplo:

```python
nomes = ["Ana", "Bruno", "Carla"]

for indice, nome in enumerate(nomes):
    print(indice, nome)
```

A cada iteração:

- `indice` recebe `0`, depois `1`, depois `2`;
- `nome` recebe `"Ana"`, depois `"Bruno"`, depois `"Carla"`.

Isso evita precisar criar e atualizar um contador manualmente.


> Este recurso não foi identificado nas aulas/exercícios considerados até 02/10/2026. Ele foi incluído porque torna mais simples acessar **posição + valor** sem manter um contador manual.


````python
nomes = ["Ana", "Bruno", "Carla"]

for indice, nome in enumerate(nomes):
    print("Índice:", indice, "| Nome:", nome)
````

### Por que existem duas variáveis depois de `for`?

`enumerate(nomes)` fornece, a cada passo, um par parecido com:

```text
(0, "Ana")
(1, "Bruno")
(2, "Carla")
```

O Python faz o **desempacotamento** desse par:

- primeiro valor → `indice`;
- segundo valor → `nome`.

É a mesma ideia de desempacotamento que vimos com tuplas no Checkpoint 2.


## 3.5 `while` — repetir enquanto uma condição for verdadeira

`while` significa **“enquanto”**.

Estrutura básica:

```python
while condição:
    # código repetido
```

O Python:

1. testa a condição;
2. se for `True`, executa o bloco;
3. volta e testa novamente;
4. repete enquanto a condição continuar `True`.

Por isso, é essencial que alguma coisa dentro do processo faça a condição **mudar**. Caso contrário, podemos criar um **loop infinito**.


````python
tentativas = 0

while tentativas < 3:
    print("Tentativa", tentativas + 1)
    tentativas += 1

print("Loop encerrado. Valor final de tentativas:", tentativas)
````

### O que significa `tentativas += 1`?

É uma forma abreviada de escrever:

```python
tentativas = tentativas + 1
```

Ou seja:

> pegue o valor atual de `tentativas`, some `1` e guarde o novo resultado na própria variável.

No exemplo:

- começa em `0`;
- depois vira `1`;
- depois `2`;
- depois `3`;
- quando `tentativas < 3` passa a ser `False`, o `while` termina.

O `+ 1` usado apenas no `print()` serve para mostrar ao usuário “Tentativa 1”, “Tentativa 2” e “Tentativa 3”, embora o contador interno comece em zero.


## 3.6 `break` e `continue` — controlar um loop

Dentro de um loop, às vezes queremos alterar o fluxo normal da repetição.

### `continue` — pule esta iteração

`continue` interrompe **somente a iteração atual** e manda o loop seguir para o próximo item.

### `break` — encerre o loop

`break` encerra **todo o loop imediatamente**. Os itens seguintes deixam de ser processados.

Compare mentalmente:

- `continue` → “este item não; vá para o próximo”;
- `break` → “pare o loop agora”.


````python
valores = [10, -1, 20, 999, 30]

for valor in valores:
    if valor < 0:
        print("Ignorando valor negativo:", valor)
        continue

    if valor > 100:
        print("Valor acima de 100 encontrado:", valor)
        print("Encerrando o loop.")
        break

    print("Valor processado:", valor)
````

### Lendo o exemplo passo a passo

Com a lista `[10, -1, 20, 999, 30]`:

1. `10` → não é negativo e não é maior que 100 → é processado;
2. `-1` → condição `valor < 0` é verdadeira → `continue` pula o restante daquela iteração;
3. `20` → é processado normalmente;
4. `999` → é maior que 100 → `break` encerra o loop;
5. `30` → **nunca é analisado**, porque o loop já terminou.

Observe que `continue` e `break` só fazem sentido **dentro de estruturas de repetição**.


### 💡 Dica de Ouro — indentação muda a lógica

Em Python, o espaço no início da linha indica **a qual bloco a instrução pertence**.

```python
for venda in vendas:
    total += venda
print(total)
```

O `print()` está fora do `for`: aparece uma vez.

```python
for venda in vendas:
    total += venda
    print(total)
```

Agora está dentro do `for`: aparece a cada iteração.

Esse tipo de diferença gerou correção prática em aula. Se uma saída repete demais — ou de menos — confira a indentação antes de mudar toda a lógica.


## 3.7 `for` + dicionário + acumulador

Quando `for` percorre uma lista de dicionários, a variável temporária recebe **um dicionário inteiro por vez**.

```python
valor_estoque = 0

for item in produtos:
    valor_item = item["preco"] * item["estoque"]
    valor_estoque += valor_item

print(f"Valor potencial do estoque: R$ {valor_estoque:.2f}")
```

Em cada iteração:

1. `item` representa um produto;
2. `item["preco"] * item["estoque"]` calcula o valor daquele produto;
3. `valor_estoque += valor_item` acumula no total.

O acumulador precisa ser inicializado **antes** do `for`. Se `valor_estoque = 0` ficar dentro do laço, o total será reiniciado a cada iteração.

### Combinar `for` e `if`

```python
for item in produtos:
    if item["estoque"] < 5:
        print(item["nome"])
```

- `for` visita todos os produtos;
- `if` decide quais atendem à regra;
- `< 5` é diferente de `<= 5`.

A interpretação da regra de negócio vem antes do operador. “Menor que cinco” exclui `5`; “cinco ou menos” inclui `5`.


## 3.8 Um mesmo laço pode calcular, classificar e registrar

A aula de 02/10 combinou contador, acumulador, condição booleana e lista de resultados:

```python
soma_notas = 0
aprovados = 0
baixa_presenca = []

for aluno in alunos:
    aluno["aprovado"] = (
        aluno["nota"] >= 7
        and aluno["presenca"] >= 75
    )

    soma_notas += aluno["nota"]

    if aluno["aprovado"]:
        aprovados += 1

    if aluno["presenca"] < 75:
        baixa_presenca.append(aluno["nome"])
```

Aqui existem três objetivos diferentes:

- `soma_notas` → **acumulador**;
- `aprovados` → **contador**;
- `baixa_presenca` → **lista de registros selecionados**.

Os dois `if` são independentes porque verificam fatos diferentes. Esse padrão é importante em análise de dados: transformar regras explícitas em condições computáveis sem perder o significado da regra.

# ✅ CHECKPOINT 4 — Funções

Uma **função** é um bloco de código que recebe um nome e pode ser executado sempre que precisarmos daquela lógica.

Funções ajudam a:

- evitar repetição de código;
- dividir um problema em partes menores;
- dar nomes claros às operações;
- reutilizar uma mesma lógica com valores diferentes.

Neste checkpoint veremos a anatomia de uma função, parâmetros, argumentos, chamada, `return`, valores padrão e múltiplos retornos.


## 4.1 `def` — definir uma função

Para criar uma função usamos `def`, abreviação de **define** (“definir”).

Exemplo:

```python
def saudacao(nome):
    return f"Olá, {nome}"
```

Leia a primeira linha como:

> “Defina uma função chamada `saudacao` que recebe uma informação chamada `nome`.”

Anatomia:

- `def` → inicia a definição da função;
- `saudacao` → nome escolhido para a função;
- `(nome)` → parâmetros que a função espera receber;
- `:` → inicia o bloco da função;
- indentação → mostra quais linhas pertencem à função;
- `return` → devolve um resultado.

Definir uma função **não significa executá-la imediatamente**. Primeiro criamos a função; depois fazemos uma chamada.


````python
def saudacao(nome):
    return f"Olá, {nome}"

print("A função foi definida. Agora vamos chamá-la:")
mensagem = saudacao("Ana")
print(mensagem)
````

### O que é uma chamada de função?

Esta parte:

```python
saudacao("Ana")
```

é uma **chamada** da função.

Os parênteses indicam que queremos executar `saudacao` e passar o texto `"Ana"` para ela.

Fluxo:

1. a chamada envia `"Ana"`;
2. dentro da função, `nome` recebe `"Ana"`;
3. a f-string produz `"Olá, Ana"`;
4. `return` devolve essa string;
5. o valor retornado é guardado em `mensagem`.


## 4.2 Parâmetro × argumento

Os dois termos são parecidos, mas descrevem momentos diferentes.

### Parâmetro

É o **nome usado na definição da função** para representar um valor que ainda será recebido.

```python
def dobro(numero):
```

`numero` é um **parâmetro**.

### Argumento

É o **valor real enviado quando chamamos a função**.

```python
dobro(5)
```

`5` é o **argumento**.

Podemos resumir assim:

```text
definição → parâmetro
chamada   → argumento
```


````python
def dobro(numero):
    return numero * 2

resultado_1 = dobro(5)
resultado_2 = dobro(12)

print("dobro de 5:", resultado_1)
print("dobro de 12:", resultado_2)
````

Observe que a lógica da função é a mesma. O que muda é o argumento enviado em cada chamada.

Isso é uma das principais vantagens das funções: **reutilizar a mesma regra com entradas diferentes**.


## 4.3 `return` × `print()`

Esses dois comandos podem mostrar resultados semelhantes durante um teste, mas têm funções diferentes.

### `print()`

Mostra uma informação na saída do notebook.

### `return`

Entrega um valor **de volta para quem chamou a função**.

Esse valor pode ser:

- guardado em uma variável;
- usado em outro cálculo;
- passado para outra função;
- exibido depois com `print()`.

Uma função que apenas usa `print()` mostra algo, mas não devolve aquele resultado como valor reutilizável.


````python
def soma_exibida(a, b):
    print("Dentro da função soma_exibida:", a + b)


def soma_retornada(a, b):
    return a + b


r1 = soma_exibida(2, 3)
r2 = soma_retornada(2, 3)

print("Valor guardado em r1:", r1)
print("Valor guardado em r2:", r2)
print("r2 ainda pode ser usado em outro cálculo:", r2 * 10)
````

### Por que `r1` vira `None`?

`soma_exibida()` executa `print()`, mas não possui um `return` explícito.

Quando uma função termina sem devolver um valor explicitamente, Python retorna `None`.

Já `soma_retornada()` devolve `a + b`, então `r2` recebe `5` e pode continuar sendo usado pelo programa.


## 4.4 Valores padrão em parâmetros **(complemento de apoio)**

Podemos definir um **valor padrão** para um parâmetro.

Exemplo:

```python
def formatar_nome(nome, prefixo="Cliente"):
    return f"{prefixo}: {nome}"
```

Aqui:

- `nome` precisa ser informado;
- `prefixo` já possui o valor padrão `"Cliente"`.

Se o argumento de `prefixo` não for enviado, o padrão será usado. Se for enviado, o novo valor substitui o padrão naquela chamada.


> Parâmetros com valor padrão não foram identificados nas aulas/exercícios considerados até 02/10/2026. O exemplo foi mantido porque amplia naturalmente a ideia de parâmetro e argumento.


````python
def formatar_nome(nome, prefixo="Cliente"):
    return f"{prefixo}: {nome}"

print(formatar_nome("Ana"))
print(formatar_nome("Ana", "Usuária"))
````

Leitura das chamadas:

```python
formatar_nome("Ana")
```

- `nome` recebe `"Ana"`;
- `prefixo` permanece com `"Cliente"`.

Já:

```python
formatar_nome("Ana", "Usuária")
```

- `nome` recebe `"Ana"`;
- `prefixo` recebe `"Usuária"` no lugar do padrão.


## 4.5 Mais de um valor retornado **(complemento de apoio)**

Uma função pode devolver mais de uma informação.

Exemplo:

```python
def extremos(valores):
    return min(valores), max(valores)
```

Aqui usamos duas funções já disponíveis no Python:

- `min(valores)` → menor valor da coleção;
- `max(valores)` → maior valor da coleção.

O `return` devolve os dois resultados juntos.


> O retorno de múltiplos valores não foi identificado nas aulas/exercícios considerados até 02/10/2026. `max()` já apareceu em atividade de análise; `min()` e o retorno conjunto aparecem aqui como extensão didática.


````python
def extremos(valores):
    return min(valores), max(valores)

resultado = extremos([4, 9, 2, 7])

print("Resultado completo:", resultado)
print("Tipo do resultado:", type(resultado))
````

O resultado é uma **tupla**:

```text
(2, 9)
```

Por isso também podemos desempacotar:

```python
menor, maior = extremos([4, 9, 2, 7])
```

- primeiro valor retornado → `menor`;
- segundo valor retornado → `maior`.


````python
menor, maior = extremos([4, 9, 2, 7])

print("Menor:", menor)
print("Maior:", maior)
````

Ao revisar uma função, pergunte:

Isso evita resolver apenas parte do problema.

## 5.1 Métodos de string

Um **método** é uma operação associada a um objeto.

Quando escrevemos:

```python
texto.upper()
```

podemos ler como:

> “No objeto `texto`, execute o método `upper()`.”

O ponto `.` conecta o objeto ao método que queremos usar.

Alguns métodos frequentes:

| Método | O que faz | Situação neste material |
|---|---|---|
| `.upper()` | transforma letras em maiúsculas | **complemento de apoio** |
| `.lower()` | transforma letras em minúsculas | **complemento de apoio** |
| `.strip()` | remove espaços em branco do início e do fim | já aparece no conteúdo de 02/10 |
| `.replace(antigo, novo)` | substitui um trecho por outro | **complemento de apoio** |

Esses métodos **retornam uma nova string**. A string original não é alterada automaticamente.


> `.upper()`, `.lower()` e `.replace()` não foram identificados nas aulas/exercícios considerados até 02/10/2026. Foram mantidos por serem operações básicas e fáceis de comparar com `.strip()`, que já aparece no conteúdo trabalhado.


````python
texto = "  Python para Dados  "

print("Original:", repr(texto))
print("strip():", repr(texto.strip()))
print("upper():", texto.upper())
print("lower():", texto.lower())
print("replace():", texto.replace("Dados", "Análise"))

print("Original depois das operações:", repr(texto))
````

### Por que usei `repr(texto)` no exemplo? **(complemento de apoio)**

`repr()` mostra uma representação mais explícita do valor. Isso ajuda a enxergar espaços que poderiam passar despercebidos.

Assim:

```text

```

mostra claramente que existem espaços antes e depois do texto.

## 5.2 Encadeamento de métodos **(complemento de apoio)**

Podemos aplicar um método e, no resultado dele, aplicar outro método.

Exemplo:

```python
texto.strip().upper()
```

Leia da esquerda para a direita:

1. `texto.strip()` → remove espaços das extremidades;
2. o resultado recebe `.upper()` → transforma as letras em maiúsculas.

Isso é chamado de **encadeamento de métodos**.


> O encadeamento explícito de métodos não foi identificado nas aulas/exercícios considerados até 02/10/2026. Ele foi incluído para mostrar como pequenas transformações podem ser compostas em uma mesma expressão.


````python
texto = "  Python para Dados  "

limpo = texto.strip()
padronizado = limpo.upper()

print("Em duas etapas:", padronizado)
print("Encadeado:", texto.strip().upper())
````

As duas formas produzem o mesmo resultado. A versão em etapas costuma ser mais fácil para aprender e depurar; a versão encadeada é mais compacta quando a leitura continua clara.


## 5.3 Regex — reconhecer padrões em texto

**Regex** é abreviação de **regular expression** (expressão regular).

Uma expressão regular descreve um **padrão de texto**, em vez de procurar apenas uma palavra exata.

Exemplos de padrões básicos usados neste corte:

| Padrão | Significado |
|---|---|
| `\d` | um dígito de `0` a `9` |
| `\D` | um caractere que **não** é dígito |

Para usar expressões regulares em Python, importamos o módulo `re`:

```python
import re
```

`import` disponibiliza no programa funcionalidades que estão organizadas em um módulo/biblioteca.


````python
import re

print("Módulo re carregado:", re)
````

### Por que aparece `r` antes das aspas?

Em regex é comum escrever:

```python
r"\D"
```

O `r` cria uma **raw string** (“string crua”). Ela ajuda a escrever barras invertidas `\` de forma mais direta, algo muito comum em expressões regulares.

Neste nível, a regra prática é:

> ao escrever padrões de regex, prefira `r"..."`.


## 5.4 `re.sub()` — substituir o que combina com um padrão

`sub` vem de **substitute** (“substituir”).

Estrutura:

```python
re.sub(padrao, substituicao, texto)
```

No exemplo:

```python
re.sub(r"\D", "", telefone)
```

leia assim:

> “No texto `telefone`, encontre todo caractere que não seja dígito (`\D`) e substitua por texto vazio (`""`).”

Na prática, isso remove parênteses, espaço e hífen, deixando somente os números.


````python
import re

telefone = "(21) 98888-1111"
somente_numeros = re.sub(r"\D", "", telefone)

print("Original:", telefone)
print("Somente números:", somente_numeros)
````

Saída esperada:

```text
21988881111
```

Observe que `re.sub()` retorna uma nova string; a variável `telefone` original continua existindo com a formatação anterior.


## 5.5 `re.match()` — verificar um padrão no início

`re.match()` tenta verificar se um padrão aparece **a partir do início da string**.

Exemplo simples:

```python
re.match(r"\d", "7 dias")
```

Como o texto começa com um dígito, há correspondência.

Se não houver correspondência, o resultado é `None`.

Neste estágio, não precisamos explorar todos os recursos de regex; o objetivo é entender a ideia de **testar um padrão textual**.


````python
import re

texto_1 = "7 dias"
texto_2 = "Sete dias"

resultado_1 = re.match(r"\d", texto_1)
resultado_2 = re.match(r"\d", texto_2)

print("texto_1 começa com dígito?", resultado_1 is not None)
print("texto_2 começa com dígito?", resultado_2 is not None)
````

## 5.6 `lambda` — função curta de uma única expressão

`lambda` permite criar uma função pequena sem usar a estrutura completa com `def`.

Compare:

```python
def dobro(x):
    return x * 2
```

com:

```python
dobro = lambda x: x * 2
```

As duas versões podem produzir o mesmo resultado para esse caso simples.

Anatomia:

```text
lambda x: x * 2
       │  └──── expressão calculada e retornada
       └─────── parâmetro
```

- `lambda` inicia a função curta;
- `x` é o parâmetro;
- `:` separa os parâmetros da expressão;
- `x * 2` é calculado e retornado automaticamente.

Não escrevemos `return` dentro dessa forma simples.


````python
def dobro_com_def(x):
    return x * 2


dobro_com_lambda = lambda x: x * 2

print("Com def:", dobro_com_def(7))
print("Com lambda:", dobro_com_lambda(7))
````

### Quando usar `lambda`?

Neste nível, use `lambda` apenas quando a operação for **muito curta e fácil de entender**.

Se a lógica precisar de várias etapas, explicações, validações ou condições mais complexas, uma função com `def` normalmente é mais legível.

Portanto, `lambda` não substitui `def`; é uma ferramenta adequada para alguns casos pequenos.
