## 🧠 Regra de leitura de código em Pandas

Há três padrões que aparecerão muitas vezes.

### 1. Atribuição: primeiro a direita, depois a esquerda

```python
precos = df["preco"]
```

Leia assim:

1. `df["preco"]` é avaliado primeiro;
2. o resultado é guardado na variável `precos`.

### 2. Encadeamento: acompanhe da esquerda para a direita

```python
df["pais"].value_counts().head(5)
```

Leia assim:

1. `df["pais"]` → selecione a coluna `pais`;
2. `.value_counts()` → conte quantas vezes cada valor aparece;
3. `.head(5)` → mostre os cinco primeiros resultados.

### 3. Função com expressão dentro: resolva a parte interna

```python
pd.to_datetime(df["data_pedido"], errors="coerce")
```

Leia assim:

1. `df["data_pedido"]` → selecione a coluna;
2. `pd.to_datetime(...)` → tente converter essa coluna para data;
3. `errors="coerce"` → valores inválidos viram `NaT` em vez de interromper a execução.

## 1.2 O que é um `DataFrame`?

Um **DataFrame** é uma estrutura tabular do Pandas.

Pense em algo semelhante a uma planilha:

| índice | produto | preco | estoque |
|---:|---|---:|---:|
| 0 | Mouse | 89.90 | 14 |
| 1 | Teclado | 249.90 | 7 |
| 2 | Webcam | 179.90 | 5 |

Temos três ideias diferentes:

- **linha** → um registro;
- **coluna** → uma característica/variável;
- **índice** → identificação das linhas dentro do DataFrame.

O índice `0, 1, 2...` não é necessariamente uma coluna da base original. Ele é uma estrutura que o Pandas usa para localizar linhas.

### 🧩 Como ler tipos como `int64`, `float64` e `object`

Aqui começa uma diferença importante entre Python puro e Pandas.

Em Python, vimos tipos como:

```python
int
float
str
bool
```

No Pandas, muitas colunas usam tipos vindos do ecossistema NumPy/Pandas, por isso você pode ver:

| Tipo mostrado | Leitura inicial |
|---|---|
| `int64` | números inteiros |
| `float64` | números com ponto flutuante |
| `bool` | verdadeiro/falso |
| `object` | tipo genérico; frequentemente texto |
| `datetime64[ns]` | data/hora reconhecida pelo Pandas |

### Por que `float64` e não apenas `float`?

`float` é o nome do tipo numérico de ponto flutuante do Python.

`float64` indica um tipo numérico usado em estruturas do NumPy/Pandas e que utiliza **64 bits** de armazenamento.

O `64` **não significa 64 casas decimais**.

Para este estágio, a leitura prática é:

> `float64` = “esta coluna está sendo tratada como número decimal”.

O mesmo raciocínio vale para `int64`:

> `int64` = “esta coluna está sendo tratada como número inteiro”.

> 🧩 **Nota de leitura:** detalhes internos de representação binária não são necessários agora. O importante é reconhecer o que o tipo permite fazer.


### Exemplo comparativo — inteiro, decimal e texto

Vamos criar uma tabela pequena para observar os tipos:


````python
tipos_exemplo = pd.DataFrame({
    "quantidade": [1, 2, 3],
    "preco": [10.50, 20.00, 35.75],
    "categoria": ["A", "B", "A"]
})

print(tipos_exemplo)
print()
print("Tipos:")
print(tipos_exemplo.dtypes)
````

O esperado é algo próximo de:

```text
quantidade     int64
preco        float64
categoria     object
```

A exibição exata pode variar conforme a versão/configuração do Pandas, principalmente para texto. O ponto principal é perceber que **cada coluna possui um tipo**.


### Exemplo comparativo — valor ausente pode alterar o tipo

Em muitas configurações tradicionais do Pandas, uma coluna de inteiros que contém `None` pode aparecer como `float64`:


````python
com_ausencia = pd.DataFrame({
    "quantidade": [10, 20, None]
})

print(com_ausencia)
print()
print("Tipo:", com_ausencia["quantidade"].dtype)
````

Por quê?

O Pandas precisa representar tanto:

- os números `10` e `20`;
- quanto o valor ausente.

Em representações tradicionais, isso pode levar a coluna para `float64`, usando um marcador numérico de ausência.

> 💡 **Dica de Ouro:** se um tipo parecer estranho, não conclua imediatamente que a base está errada. Primeiro investigue **os valores e as ausências da coluna**.

## 4.1 Valores ausentes com `isna().sum()`

```python
df.isna().sum()
```

Leia da esquerda para a direita:

1. `df` → DataFrame;
2. `.isna()` → para cada célula, pergunte “está ausente?”;
3. o resultado contém `True` ou `False`;
4. `.sum()` → some os `True` de cada coluna.

Em uma soma, `True` funciona como `1` e `False` como `0`. Por isso obtemos a **quantidade de ausências por coluna**.


````python
df_qualidade = pd.DataFrame({
    "produto": ["Mouse", "Mouse", "Teclado", "Webcam"],
    "preco": [89.90, 89.90, None, 179.90],
    "estoque": [14, 14, 7, 5]
})

print(df_qualidade)

print()
print("Mapa True/False de ausências:")
print(df_qualidade.isna())

print()
print("Quantidade por coluna:")
print(df_qualidade.isna().sum())
````

### Comparação importante

```python
df.isna()
```

responde **onde** existem ausências.

```python
df.isna().sum()
```

responde **quantas** ausências existem em cada coluna.


## 4.2 Duplicidades: detectar antes de remover

```python
df.duplicated()
```

marca linhas repetidas.

```python
df.duplicated().sum()
```

conta quantas linhas foram marcadas como duplicadas.

```python
df.drop_duplicates()
```

produz uma versão sem as duplicatas consideradas.

Mas duas linhas iguais não são automaticamente erro. Duas compras iguais podem ter acontecido de verdade.


````python
print("Marcador de duplicidade:")
print(df_qualidade.duplicated())

print()
print("Quantidade duplicada:")
print(df_qualidade.duplicated().sum())

print()
print("Resultado sem duplicatas:")
print(df_qualidade.drop_duplicates())
````

## 4.3 `copy()` — preservar uma referência de trabalho

Na aula de 02/10 apareceu:

```python
df = df_csv.copy()
```

Leia:

1. `df_csv` → base carregada;
2. `.copy()` → crie uma cópia em memória;
3. `df =` → guarde essa cópia em `df`.

A partir daí, podemos tratar `df` e ainda ter `df_csv` para comparação.

### Importante

`copy()` **não cria um novo CSV, JSON ou Excel no disco**.

Ela cria outro objeto em memória.

Salvar arquivo é outra etapa, feita com `to_csv()`, `to_json()` ou `to_excel()`.


````python
original = df_qualidade.copy()
tratado = original.drop_duplicates().copy()

print("Original:")
print(original)

print()
print("Tratado:")
print(tratado)
````

## 5.2 Padronizar categorias com `replace()`

Imagine:


Para o computador, são três strings diferentes.

Podemos criar um mapa:


````python
paises = pd.DataFrame({
    "pais": ["Brasil", "Brazil", "BR", "Argentina", "BR"]
})

mapa_paises = {
    "Brasil": "Brasil",
    "Brazil": "Brasil",
    "BR": "Brasil"
}

paises["pais_padronizado"] = paises["pais"].replace(mapa_paises)

print(paises)
````

## 5.3 `apply()` — aplicar uma regra valor por valor

```python
df["email"].apply(...)
```

`apply()` recebe uma função e a executa para cada valor da coluna.

Podemos fazer isso com uma função nomeada:


````python
emails = pd.DataFrame({
    "email": [" ana@email.com ", "bruno@email.com", " carla@email.com  "]
})

def remover_espacos(texto):
    return str(texto).strip()

emails["com_funcao"] = emails["email"].apply(remover_espacos)
print(emails)
````

Ou com uma `lambda` curta:


````python
emails["com_lambda"] = emails["email"].apply(
    lambda x: str(x).strip()
)

print(emails)
````

Compare:

```python
.apply(remover_espacos)
```

e:

```python
.apply(lambda x: str(x).strip())
```

A ideia é a mesma: entregar ao `apply()` uma regra que será aplicada aos valores.

A `lambda` é útil quando a regra é curta. Se a lógica crescer, uma função com `def` costuma ser mais legível.


## 5.4 `DataFrame.apply(..., axis=1)` — aplicar uma regra por linha

Na aula de 22/09, `apply()` também apareceu no **DataFrame inteiro**, com `axis=1`.

Compare:

```python
df["email"].apply(funcao)
```

Aqui a função recebe **um valor por vez de uma Series**.

Já:

```python
df.apply(funcao, axis=1)
```

com `axis=1`, a função recebe **uma linha por vez**. Isso é útil quando o cálculo depende de **mais de uma coluna da mesma linha**.

Exemplo:


````python
pedidos = pd.DataFrame({
    "produto": ["Mouse", "Teclado", "Webcam"],
    "preco": [89.90, 249.90, 179.90],
    "quantidade": [2, 1, 3],
})

pedidos["valor_item"] = pedidos.apply(
    lambda linha: linha["preco"] * linha["quantidade"],
    axis=1,
)

print(pedidos)
````

Leia a expressão principal:

1. `pedidos.apply(...)` → aplique uma função ao DataFrame;
2. `axis=1` → entregue **cada linha** à função;
3. `lambda linha: ...` → a variável `linha` representa o registro atual;
4. `linha["preco"] * linha["quantidade"]` → combine duas colunas daquele mesmo registro;
5. o resultado de cada linha forma a nova coluna `valor_item`.

> 💡 **Dica de leitura:** `axis=1` não significa “coluna 1”. Neste uso de `DataFrame.apply`, ele indica **aplicação por linha**.

# ✅ CHECKPOINT 7 — Datas, durações e interpretação

Datas parecem texto quando chegam de alguns arquivos, mas precisamos convertê-las para um tipo de data antes de fazer operações de calendário com segurança.


## 7.1 Texto de data × data reconhecida

Considere:

```python
"2026-10-02"
```

Visualmente parece uma data, mas entre aspas é uma **string**.

Depois de conversão, o Pandas pode reconhecer a coluna como algo parecido com:

```text
datetime64[ns]
```

Para o nosso nível:

> `datetime64[ns]` = “Pandas reconheceu esses valores como data/hora”.

O trecho `[ns]` indica uma resolução interna baseada em nanossegundos. Você **não precisa memorizar isso** para usar datas corretamente.


## 7.2 Uma data isolada com `datetime.strptime()`

```python
from datetime import datetime

texto = "02/10/2026"
data = datetime.strptime(texto, "%d/%m/%Y")
```

Leia:

1. `texto` contém uma string;
2. `datetime.strptime(...)` tenta interpretar essa string;
3. `"%d/%m/%Y"` descreve o formato esperado.

| Código | Significado |
|---|---|
| `%d` | dia |
| `%m` | mês |
| `%Y` | ano com quatro dígitos |


````python
from datetime import datetime

texto = "02/10/2026"
data = datetime.strptime(texto, "%d/%m/%Y")

print("Antes:", texto, type(texto))
print("Depois:", data, type(data))
````

## 7.3 Uma coluna inteira com `pd.to_datetime()`

Para uma coluna:

```python
df["data_pedido"] = pd.to_datetime(
    df["data_pedido"],
    format="%Y-%m-%d",
    errors="coerce",
)
```

Leia em blocos:

1. `df["data_pedido"]` → coluna original;
2. `pd.to_datetime(...)` → converta para data;
3. `format="%Y-%m-%d"` → formato esperado;
4. `errors="coerce"` → valores inválidos viram `NaT`;
5. `df["data_pedido"] =` → substitua a coluna pelo resultado convertido.


## 7.4 `errors="coerce"` e `NaT`

Veja um exemplo com uma data inválida:


````python
datas = pd.DataFrame({
    "texto_data": ["2026-10-01", "data_invalida", "2026-10-03"]
})

datas["data_convertida"] = pd.to_datetime(
    datas["texto_data"],
    format="%Y-%m-%d",
    errors="coerce",
)

print(datas)
print()
print("Tipos:")
print(datas.dtypes)
````

O valor inválido vira `NaT`.

`NaT` significa, de forma prática:

> **Not a Time** — data/hora ausente ou não convertida.

Ele é semelhante à ideia de ausência, mas específico para dados temporais.

### O que `errors="coerce"` NÃO faz

Ele não “corrige” magicamente a data.

Ele impede que um valor inválido interrompa toda a conversão e o marca para investigação.


## 7.5 Extrair partes da data com `.dt`

Depois que a coluna é temporal:

```python
df["data_pedido"].dt.year
df["data_pedido"].dt.month
df["data_pedido"].dt.day_name()
```

`.dt` é um **acessor**: ele abre operações específicas para dados de data/hora.

Pense assim:

```text
coluna de data → .dt → componente desejado
```


````python
datas_validas = pd.DataFrame({
    "data": pd.to_datetime(["2026-10-01", "2026-10-02", "2026-10-03"])
})

datas_validas["ano"] = datas_validas["data"].dt.year
datas_validas["mes"] = datas_validas["data"].dt.month
datas_validas["dia_semana"] = datas_validas["data"].dt.day_name()

print(datas_validas)
````

## 7.6 Diferença entre datas e `.dt.days`

Vamos criar duas datas por pedido:


````python
pedidos = pd.DataFrame({
    "data_pedido": ["2026-09-01", "2026-09-05", "2026-09-10"],
    "data_entrega": ["2026-09-04", "2026-09-08", None],
})

pedidos["data_pedido"] = pd.to_datetime(
    pedidos["data_pedido"],
    errors="coerce",
)
pedidos["data_entrega"] = pd.to_datetime(
    pedidos["data_entrega"],
    errors="coerce",
)

pedidos["duracao"] = pedidos["data_entrega"] - pedidos["data_pedido"]
pedidos["prazo_dias"] = pedidos["duracao"].dt.days

print(pedidos)
````

Leia a expressão:

```python
(pedidos["data_entrega"] - pedidos["data_pedido"]).dt.days
```

1. selecione `data_entrega`;
2. selecione `data_pedido`;
3. subtraia uma data da outra;
4. o resultado é uma duração;
5. `.dt.days` extrai a quantidade de dias.

Se a data de entrega estiver ausente, a duração também fica ausente.


## 7.7 Prazo observado não é atraso automaticamente

Se calculamos:

```text
pedido: 01/09
entrega: 04/09
duração: 3 dias
```

sabemos que o processo levou 3 dias.

Mas isso **não prova atraso**.

Para afirmar atraso, precisamos de outra informação, por exemplo:

```text
prazo prometido: 2 dias
```

Aí sim conseguimos comparar:

```text
3 dias realizados > 2 dias prometidos
```

### Mais duas regras importantes

- data ausente não autoriza inventar uma data;
- duração desconhecida não deve virar `0`, porque `0` significa “ocorreu no mesmo dia”.

> 💡 O computador calcula o que está disponível. O analista decide **o que o cálculo permite concluir**.
