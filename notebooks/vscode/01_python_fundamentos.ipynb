# ✅ CHECKPOINT 2 — Valores e tipos

## 2.1 O tipo muda o que o Python pode fazer

Um valor pode ser texto, número inteiro, número decimal, verdadeiro/falso ou ausência de valor.

```python
"42"     # str

42.0     # float

None     # NoneType
```

Esses valores podem parecer semelhantes visualmente, mas se comportam de formas diferentes.


````python
valores = ["42", 42, 42.0, True, None]
for valor in valores:
    print(repr(valor), "->", type(valor))
````


junta dois textos e produz `"428"`.

```python

```

soma números e produz `50`.

As aspas mudam o **tipo** do valor.


````python
print("42" + "8")
print(42 + 8)
````

## 2.3 `int` e `float`

- `int`: número inteiro, sem parte decimal.
- `float`: número com parte decimal.


## 2.5 `None`: ausência de valor

`None` não é um espaço em branco, não é zero e não é o texto `"None"`.

Compare:

```python
a = None      # ausência de valor
b = ""        # string vazia
c = "None"    # texto com quatro caracteres
d = "."       # texto com um caractere
```

Se você substituir `None` por `"."`, a variável passa a conter uma **string**.

Quando fazemos:

```python

```

a palavra `None` aparece na tela porque essa é a representação textual do objeto. Isso **não** transforma o valor em string.


> 🧩 **Complemento de apoio — `repr()`**: no exemplo executável abaixo, `repr()` é usado apenas para deixar visíveis detalhes como aspas e espaços. Esse recurso não foi identificado nas aulas/exercícios considerados até 02/10/2026; ele aparece aqui como ferramenta de leitura e diagnóstico.


````python
a = None
b = ""
c = "None"
d = "."

for nome, valor in [("a", a), ("b", b), ("c", c), ("d", d)]:
    print(nome, "=", repr(valor), "| tipo:", type(valor))
````

## 2.6 `type()` revela o tipo


Podemos ler assim:

`type()` é uma ferramenta de diagnóstico muito útil quando não sabemos como o Python está interpretando um valor.


````python
valores = ["42", 42, 42.0, True, None]

for valor in valores:
    print(repr(valor), "->", type(valor))
````

````python
preco_direto = float(input("Digite outro preço: "))

print("Valor final:", preco_direto)
print("Tipo final:", type(preco_direto))
````

# ✅ CHECKPOINT 4 — Cálculo e apresentação

## 4.1 Operadores aritméticos básicos

Os operadores não fazem todos a mesma “conta”. A diferença entre `/`, `//` e `%` é especialmente importante:

| Operador | Nome | O que faz | Exemplo |
|---|---|---|---|
| `+` | adição | soma valores | `10 + 3` → `13` |
| `-` | subtração | calcula a diferença | `10 - 3` → `7` |
| `*` | multiplicação | multiplica valores | `10 * 3` → `30` |
| `/` | divisão | calcula o quociente e pode manter a parte decimal | `10 / 3` → `3.333...` |
| `//` | divisão inteira | calcula quantos grupos inteiros completos cabem | `10 // 3` → `3` |
| `%` | módulo / resto | devolve o que sobra da divisão inteira | `10 % 3` → `1` |
| `**` | potenciação | eleva um número a uma potência | `2 ** 3` → `8` |

### `/` × `//`: por que o resultado muda?

Com `10 / 3`, perguntamos pelo **resultado da divisão**, incluindo a parte decimal.

Com `10 // 3`, perguntamos quantos grupos **inteiros completos** de 3 cabem em 10.

Ainda sobra `1`; essa sobra é obtida com `10 % 3`.

Uma aplicação comum de `%` é verificar paridade:

```python
numero % 2 == 0
```

Se o resto da divisão por 2 for zero, o número é par.

> 💡 **Dica de Ouro:** `/`, `//` e `%` podem usar os mesmos números, mas respondem perguntas diferentes: **resultado da divisão**, **quantos grupos inteiros cabem** e **quanto sobra**.


````python
a = 10
b = 3

print("adição:", a + b)
print("subtração:", a - b)
print("multiplicação:", a * b)
print("divisão /:", a / b)
print("divisão inteira //:", a // b)
print("resto %:", a % b)
print("potência **:", 2 ** 3)

numero = 8
print("8 é par?", numero % 2 == 0)
````

## 4.2 `\n` — quebra de linha dentro de uma string

A sequência `\n` representa uma **quebra de linha** dentro do texto.

```python
print("Nome: Ana\nIdade: 28")
```

A saída aparece em duas linhas:

```text
Nome: Ana
Idade: 28
```

`\n` é escrito **dentro das aspas** e faz parte da string.

Também pode aparecer em f-strings:

```python
nome = "Ana"
idade = 28
print(f"Nome: {nome}\nIdade: {idade}")
```

É útil para organizar uma saída sem precisar escrever vários `print()`.


````python
nome = "Ana"
idade = 28

print(f"Nome: {nome}\nIdade: {idade}")
````

## 4.3 F-string: colocar valores dentro de um texto

```python
nome = "Ana"
print(f"Olá, {nome}")
```

Anatomia:

```text
f"Olá, {nome}"
│        └── expressão avaliada pelo Python
└─────────── indica uma string formatada
```

Sem o `f`:

```python
"Olá, {nome}"
```

`{nome}` é apenas texto literal.

> **Importante:** nem toda chave `{}` exige `f`. Chaves também são usadas em estruturas como dicionários. O `f` é necessário quando queremos que `{...}` seja interpretado **dentro de uma f-string**.


````python
nome = "Ana"
print("Olá, {nome}")
print(f"Olá, {nome}")
````

## 4.4 `round(valor, 2)` × `:.2f`

Esses recursos podem produzir números visualmente parecidos, mas têm objetivos diferentes.

```python
valor = 10.5
round(valor, 2)
```

`round()` trabalha com o **valor numérico arredondado**.

```python
f"{valor:.2f}"
```

`:.2f` controla a **forma de exibição dentro da string**, exigindo duas casas decimais.


> 🧩 **Complemento de apoio — `round()` do Python**: o arredondamento tabular com `.round(2)` já aparece nos exercícios com Pandas, mas a função embutida `round(valor, 2)` não foi identificada nas atividades/aulas consideradas até 02/10/2026. Ela foi incluída aqui para separar claramente **arredondamento numérico** de **formatação de exibição**.


````python
valor = 10.5

print("round:", round(valor, 2))
print("formatação:", f"{valor:.2f}")
````

### Um exemplo monetário

```python
salario = 4500.567
print(f"Salário: R$ {salario:.2f}")
```

Não é necessário fazer:

```python
print(f"Salário: R$ {round(salario, 2):.2f}")
```

quando a finalidade é apenas apresentar duas casas decimais. O `:.2f` já resolve a apresentação.


````python
salario = 4500.567
print("valor numérico arredondado:", round(salario, 2))
print(f"apresentação monetária: R$ {salario:.2f}")
````

- por que `f"..."` continua sendo uma string;
- para que servem as chaves dentro da f-string;
- diferença conceitual entre `round(valor, 2)` e `f"{valor:.2f}"`;
- por que `:.2f` foi estudado **depois** de strings, variáveis e números.
