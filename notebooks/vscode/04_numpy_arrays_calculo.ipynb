## ✅ CHECKPOINT 1 — O que é um array? [AULA 15/16; NOTEBOOK PROF]

Uma lista Python é uma sequência genérica; um `ndarray` do NumPy representa uma coleção com forma (`shape`) e tipo (`dtype`), favorecendo operações numéricas em blocos. Não confunda **posição** de um elemento com seu **valor**: a primeira posição é índice 0. O NumPy normalmente armazena elementos com um tipo comum, podendo converter números inteiros e decimais para um tipo adequado.

**Ler a sintaxe:** `np.array([2, 4, 6])` cria array unidimensional de três elementos; `len` mede o tamanho do primeiro eixo; `.size` conta elementos totais; `.ndim` conta dimensões; `.shape` informa comprimento dos eixos; `.dtype` informa tipo dos elementos.

**Dica de Ouro:** `lista * 2` repete elementos, enquanto `array * 2` multiplica seus valores. Execução sem erro não garante significado certo.

```python
import numpy as np

lista = [2, 4, 6]
array = np.array(lista)
print("lista * 2 =", lista * 2)
print("array * 2 =", array * 2)
print("ndim =", array.ndim, "| shape =", array.shape,
      "| size =", array.size, "| dtype =", array.dtype)
```

### 1.1 Arrays 1D e 2D [AULA 16/10]

Uma matriz de 3 linhas × 4 colunas tem `shape == (3,4)`, `ndim == 2` e `size == 12`. Linhas e colunas têm significado determinado pelo problema: no exemplo de notas, cada linha é uma pessoa *fictícia* e cada coluna é uma avaliação. Arrays também podem ter três dimensões (eixos adicionais), mas **[COMPLEMENTO DE APOIO]** essa generalização não é pré-requisito para o case de notas.

Ao criar uma matriz, faça os comprimentos das linhas iguais; listas internas irregulares podem não formar um array numérico bidimensional útil.

```python
matriz_ex = np.array([[7., 8., 6., 9.], [6., 7., 8., 7.], [8., 9., 9., 8.]])
print(matriz_ex)
print("ndim:", matriz_ex.ndim, "shape:", matriz_ex.shape,
      "size:", matriz_ex.size, "dtype:", matriz_ex.dtype)
assert matriz_ex.shape == (3, 4)
```

## ✅ CHECKPOINT 2 — Vetorização, operadores e máscaras [AULA 16/10; NOTEBOOK PROF]

**Vetorização** é aplicar operações sobre o array inteiro sem escrever um `for` explícito para cada elemento. Em `notas + 0.5`, cada valor recebe acréscimo 0.5. Uma expressão `notas >= 7` produz um array booleano, posição a posição; com ele, é possível selecionar apenas os elementos que satisfazem um critério. **A máscara precisa ser compatível com a forma do dado.**

Funções como `np.mean`, `np.max`, `np.min` resumem dados; uma média geral não informa como cada linha se comportou individualmente.

```python
notas = np.array([5.5, 6.0, 7.0, 8.5])
bonus = notas + 0.5
mask = bonus >= 7
print("antes:", notas)
print("depois:", bonus)
print("máscara:", mask)
print("atendem à condição:", bonus[mask])
print("média geral:", np.mean(bonus))
```

### 2.1 Tipos, arredondamento e dados ausentes [COMPLEMENTO DE APOIO]

Nem todo valor importado é imediatamente numérico. Um array de strings não aceita todas as operações de média; `np.nan` representa um marcador numérico de ausência no contexto NumPy. `np.mean([2, np.nan])` devolve `nan`, enquanto `np.nanmean` ignora esse marcador. Ignorar ausências **é uma decisão analítica**, que exige verificar quantidade e padrão das faltas. A distinção complementa a aula de qualidade de dados, não substitui seu contexto.

Arredondar apenas para apresentar o resultado evita perder precisão intermediária. Não confundir `round` de escalar com o `np.round` em arrays.

```python
vetor_incompleto = np.array([5., np.nan, 8.])
print("missing:", np.isnan(vetor_incompleto))
print("média normal:", np.mean(vetor_incompleto))
print("média ignorando nan:", np.nanmean(vetor_incompleto))
print("quantas ausências:", np.isnan(vetor_incompleto).sum())
```

### 2.2 Operações derivadas [COMPLEMENTO DE APOIO]

`np.arange(início, fim, passo)` cria sequência com **fim exclusivo**, como `range` do Python; `np.linspace(início, fim, quantidade)` produz pontos espaçados que incluem os extremos por padrão. `np.zeros` e `np.ones` inicializam estruturas com forma definida. São recursos de apoio para compreensão e não evidência de que já foram ensinados nas gravações de 15 e 16/10.

```python
print("arange:", np.arange(0, 10, 2))
print("linspace:", np.linspace(0, 1, 5))
print("zeros:", np.zeros((2, 3)))
```

## ✅ CHECKPOINT 3 — Matrizes, indexação e eixos [AULA 16/10; NOTEBOOK PROF]

Para uma matriz **pessoas × avaliações**, o endereço `matriz[1,2]` aponta a **segunda linha** e **terceira coluna**. `matriz[:,1]` seleciona todos da segunda coluna; `matriz[0,:]` seleciona todas as avaliações da primeira pessoa. Em fatias, o índice final é exclusivo.

Uma agregação com `axis=0` **reduz o eixo das linhas**, produzindo um resultado por coluna; com `axis=1` **reduz o eixo das colunas**, produzindo um resultado por linha. A memorização “vertical/horizontal” isolada é ambígua: verifique `shape` do resultado.

```python
matriz_notas = np.array([[8., 7., 9.], [5., 6., 8.], [7., 7., 7.], [9., 8., 10.]])
print("forma:", matriz_notas.shape)
print("segunda linha, terceira coluna:", matriz_notas[1, 2])
print("primeira coluna:", matriz_notas[:, 0])
print("primeira linha:", matriz_notas[0, :])
print("média por avaliação (axis=0):", matriz_notas.mean(axis=0))
print("média por pessoa (axis=1):", matriz_notas.mean(axis=1))
assert matriz_notas.mean(axis=0).shape == (3,)
assert matriz_notas.mean(axis=1).shape == (4,)
```

### 3.1 Não confunda médias, máximos e mínimos [AULA 16/10]

`np.mean(..., axis=1)` para matriz de notas da pessoa retorna uma média por estudante; `np.min(...,axis=1)` a menor avaliação de cada estudante; `np.max` a maior. A pergunta de validação é sempre **quantos resultados espero e o que cada um significa?**

**Dica de Ouro:** “o código roda” não significa que `axis` foi escolhido corretamente. Uma transposição acidental muda a interpretação sem necessariamente causar erro.

```python
print("mínimos por pessoa:", matriz_notas.min(axis=1))
print("máximos por pessoa:", matriz_notas.max(axis=1))
print("média geral de todas as notas:", matriz_notas.mean())
```

## ✅ CHECKPOINT 4 — Broadcasting e compatibilidade de formatos [AULA 16/10; NOTEBOOK PROF]

**Broadcasting** permite combinar arrays de formas diferentes, quando são dimensionalmente compatíveis, sem repetir explicitamente todo o vetor. As dimensões são comparadas **da direita para a esquerda**: precisam ser iguais ou uma delas deve ser 1. Uma matriz de vendas de forma `(2,3)` pode ser multiplicada por um vetor `(3,)` de fatores por produto, ou um escalar.

**Risco essencial:** NumPy alinha **posição**, não nome do produto. Se a ordem dos fatores não corresponder à ordem das colunas, o resultado estará matematicamente bem formado e semanticamente errado.

```python
vendas = np.array([[10., 20., 5.], [8., 7., 12.]])  # 2 dias × 3 produtos
fatores_produto = np.array([1.10, 0.95, 1.20])
ajustado = vendas * fatores_produto
print("matriz:", vendas.shape, "fatores:", fatores_produto.shape)
print("ajustado:", ajustado)
print("totais por produto:", ajustado.sum(axis=0))
```

### 4.1 Duas direções de broadcasting [AULA 16/10; COMPLEMENTO DE APOIO]

Para adicionar um bônus específico por **linha** (por dia), transforme um vetor `(2,)` em coluna `(2,1)` via `[:, None]`. Para fator por **produto** preserve `(3,)`. Tentar combinar `(2,3)` e `(2,)` normalmente gera `ValueError` — é incompatibilidade estrutural, não erro a esconder.

**[COMPLEMENTO DE APOIO]** `reshape` pode reorganizar eixos quando a quantidade total de elementos é compatível; por si só, não corrige associação semântica por nome.

```python
bonus_por_dia = np.array([2., 3.])[:, None]
print("bonus shape:", bonus_por_dia.shape)
print(vendas + bonus_por_dia)
# Experimento de diagnóstico: tente vendas + np.array([2., 3.]) e interprete o ValueError.
```

## ✅ CHECKPOINT 5 — Integração com Pandas, decisões e interpretação [AULA 16/10; NOTEBOOK PROF]

Pandas preserva nomes de linhas/colunas e permite agregar resultados em tabela; NumPy opera eficientemente sobre arrays. Para dados tabulares de notas, primeiro selecione **explicitamente as colunas numéricas de interesse**, depois converta para array, calcule medidas com `axis=1` e reintegre ao DataFrame na mesma ordem de linhas. `np.where(condicao, valor_se_verdadeiro, valor_se_falso)` faz seleção condicional vetorizada.

No minicase da professora, uma regra de classificação ilustrativa usa **média >= 7: Aprovado; 5 <= média < 7: Revisão; < 5: Reprovado**. Essa regra pertence ao **exemplo didático da fonte**, não a critérios reais de aprovação do estudante na SC TEC. **Nomes em `notas_turma.csv` representam registros de exercício, não nota institucional individual.**

```python
import pandas as pd

tabela_sint = pd.DataFrame({
    "aluno_ficticio": ["Pessoa A", "Pessoa B", "Pessoa C", "Pessoa D"],
    "prova1": [8., 6., 4., 9.],
    "prova2": [7., 7., 4., 8.],
    "prova3": [9., 6., 5., 8.]
})
colunas = ["prova1", "prova2", "prova3"]
notas_array = tabela_sint[colunas].to_numpy(dtype=float)
media_pessoa = np.mean(notas_array, axis=1)
tabela_sint["media"] = media_pessoa
tabela_sint["situacao_didatica"] = np.where(
    media_pessoa >= 7, "Aprovado (exemplo)",
    np.where(media_pessoa >= 5, "Revisão (exemplo)", "Reprovado (exemplo)"))
print(tabela_sint.to_string(index=False))
```

### 5.1 Interpretar e controlar integridade

- A extração preserva a ordem atual das linhas. Se a tabela for reordenada entre calcular o array e devolver médias, os valores poderão ser associados à pessoa errada. Prefira atualizar imediatamente ou preservar um identificador estável e testar a correspondência.
- `value_counts` mostra categorias observadas; [AULA 16] `reindex` permite exibir também uma categoria sem ocorrências, com `fill_value=0` **apenas em contagens**.
- Uma média da turma acima de 7 não implica que todos os alunos tenham atingido 7.
- `np.where` implementa regra predeterminada; não determina sozinho qual regra de decisão é correta.

```python
categorias = ["Aprovado (exemplo)", "Revisão (exemplo)", "Reprovado (exemplo)"]
contagem = tabela_sint["situacao_didatica"].value_counts().reindex(categorias, fill_value=0)
print(contagem)
print("média geral das avaliações:", round(float(notas_array.mean()), 2))
assert contagem.sum() == len(tabela_sint)
```

### 5.2 Valores ausentes e cópia de trabalho [COMPLEMENTO DE APOIO]

Dados reais frequentemente incluem `NaN`, strings numéricas e inconsistências. Antes de `to_numpy(dtype=float)`, inspecione `.isna().sum()` e `.dtypes`, defina a política de ausências e justifique qualquer exclusão. `np.nanmean` pode ser útil apenas quando a exclusão dos valores ausentes for metodologicamente aceitável; o denominador pode variar de pessoa para pessoa. `DataFrame.copy()` ajuda a preservar um ponto de referência de trabalho.

```python
copia = tabela_sint.copy()
print(copia[colunas].isna().sum())
print("colunas:", list(copia[colunas].columns))
print("linhas:", len(copia), "shape notas:", notas_array.shape)
```
