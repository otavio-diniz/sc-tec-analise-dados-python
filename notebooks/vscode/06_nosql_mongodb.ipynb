## ✅ CHECKPOINT 5 — Arrays, $elemMatch, $unwind

[AULA M1S09–10] Filtrar `itens.produto` encontra documento com P01 mas preserva seu array completo com P02. `$elemMatch` exige múltiplas condições num mesmo elemento; `$unwind` desdobra array em linhas de pipeline, mudando granularidade. A célula não executa os operadores MongoDB.

### Arrays: correspondência em documento não é correspondência no mesmo elemento

`{"itens.produto":"P01"}` encontra **dois documentos** e preserva os arrays originais; o pedido 101 continua com P01 **e** P02. Já `{itens:{$elemMatch:{produto:"P01",qtd:{$gte:2}}}}` exige as duas condições no **mesmo item**. `{$unwind:"$itens"}` expande os arrays no pipeline e muda a granularidade: **2 pedidos → 3 itens**.

A docente trabalha esses contrastes nas aulas de 05/06 e 10/11. Nunca deduza a contagem de itens contando documentos retornados por `find`.

```python
r=[p for p in pedidos if any(i['produto']=='P01' for i in p['itens'])];assert len(r)==2 and len(r[0]['itens'])==2;print('Documentos:',len(r),'| itens pedido 101:',len(r[0]['itens']))
```

## ✅ CHECKPOINT 7 — aggregation e centavos

[AULA M1S09–10] Pipeline usa $match/$unwind/$group para selecionar, expandir e resumir. O exemplo usa centavos inteiros, mas produção exige política monetária e tipos apropriados (Decimal128). Valor previsto por pedido: 4000 e 1350 centavos, total 5350; não chamar automaticamente de receita recebida.

### Pipeline de agregação: leia etapa por etapa

`aggregate([{$match:{"cliente.estado":{$in:["SP","RJ"]}}},{$unwind:"$itens"},{$group:{_id:"$_id",unidades:{$sum:"$itens.qtd"}}},{$sort:{_id:1}}])` significa filtrar pedidos → expandir itens → reagrupar por pedido → ordenar. A saída fictícia tem 3 unidades no pedido 101 e 3 no 102. O notebook calcula a previsão em Python; o script de treino contém a consulta na linguagem do `mongosh`.

```python
v=[sum(i['qtd']*i['centavos']*(100-i['desconto'])//100 for i in p['itens']) for p in pedidos];assert v==[4000,1350];print('Centavos por pedido:',v)
```

## ✅ CHECKPOINT 8 — Modelagem e comparabilidade

[AULA M1S09–11; COMPLEMENTO] Embedding aninha cliente em pedido e pode duplicar nomes; referência separa entidades mas requer associação. ### Documentos, embedding, referências e contagens

`embedding` coloca cliente/itens dentro do pedido e favorece leitura conjunta, mas pode duplicar nome/estado; `referência` conserva ID e exige associação com cadastro em outra coleção. Na base fictícia há **3 clientes cadastrados**, **2 com pedido**, **2 documentos de pedido** e **3 itens**. Nenhuma dessas contagens é intercambiável. O `_id` Mongo não prova ser o ID de negócio; e as famílias de CSV A/B da professora têm esquemas distintos, preservados fora do repositório público.

```python
ids={c['_id'] for c in clientes};com_pedido={p['cliente']['id'] for p in pedidos};assert ids-com_pedido=={3};print('Sem pedido:',ids-com_pedido)
```
