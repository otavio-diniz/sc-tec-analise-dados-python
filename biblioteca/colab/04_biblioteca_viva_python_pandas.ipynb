**O que é:** estrutura de decisão

**Leia assim:** se / senão se / senão

**Quando aparece:** escolher caminhos

**Erro comum:** indentação incorreta ou condições sobrepostas

**Conexão:** expressão condicional

**Exemplo**

```python
if valor >= 200:
    desconto = valor * 0.10
else:
    desconto = 0
```

**O que é:** forma compacta de escolher entre dois resultados

**Leia assim:** “resultado se verdadeiro IF condição ELSE resultado se falso”

**Quando aparece:** decisões curtas

**Erro comum:** usar em regras longas e perder legibilidade

**Conexão:** if/else e lambda

**Exemplo**

```python
desconto = valor * 0.10 if valor >= 200 else 0
```


## Ambiente — Google Colab

### `célula Markdown` × `célula de código`
Markdown explica; código executa Python. A ordem de execução importa porque as variáveis vivem na sessão atual.

### sessão / reinício
Se a sessão for reiniciada ou desconectada, variáveis e arquivos temporários podem precisar ser recriados/recarregados.

### `files.upload()`
```python
from google.colab import files
uploaded = files.upload()
print(list(uploaded.keys()))
```
`files.upload()` envia arquivos para a sessão. `uploaded.keys()` mostra os nomes exatamente disponíveis; o nome usado em `pd.read_csv(...)` precisa corresponder.

### persistência
Salvar um arquivo na sessão do Colab não significa, por si só, que ele foi persistido fora daquela sessão. Diferencie **objeto em memória**, **arquivo gerado na sessão** e **arquivo armazenado em destino persistente**.
