O nome `total_vendas` não determina o tipo. **(Complemento de apoio):** `Decimal` é indicado para cálculos decimais exatos quando apropriado; `float` não garante exatidão financeira.

**💡 Dica de Ouro:** o ponto em um literal numérico define a notação decimal de Python, não a exatidão aritmética do cálculo.

## Questão 03 — Lógica com `and`

### Enunciado

**Contextualização:** Um sistema de relatório de dados precisa validar se um usuário possui a idade mínima.

**Comando:** O resultado da expressão lógica composta porresultado = (10 > 5) and (8 < 3) é definido como:

- **a)** False, visto que ambas as condições precisam ser verdadeiras.
- **b)** True, pois a primeira condição valida o restante da operação.
- **c)** None, dado que o Python não consegue comparar números de escalas diferentes.
- **d)** Erro de Sintaxe, por falta de parênteses adicionais no operador central.
- **e)** Zero, por ser o valor numérico correspondente a uma negação lógica.

> **Primeira escolha:** marque uma alternativa provisória. Em seguida, leia o laboratório e experimente os exemplos antes de revelar o comentário de resolução.

### Antes de conferir — entenda as opções

**Comparação x lógica:** `>` e `<` produzem resultados booleanos quando comparam números. `and` só satisfaz uma expressão quando **todos** os operandos booleanos são verdadeiros; `or` aceita **pelo menos um**. `None` não aparece automaticamente como resultado de comparação.

**💡 Dica de Ouro:** analise primeiro cada condição, depois o conector lógico.

### Laboratório guiado

**Investigue:** calcule cada comparação separadamente antes de avaliar a expressão completa. Troque `and` por `or`: o resultado muda?

**A seguir:** execute a célula de código, observe e explique a saída; se desejar, altere os valores.

```python
resultado = (10 > 5) and (8 < 3)
print('Condição 1:', 10 > 5)
print('Condição 2:', 8 < 3)
print('Resultado:', resultado, 'Tipo:', type(resultado).__name__)
for esquerda in (False, True):
    for direita in (False, True):
        print(esquerda, 'and', direita, '=', esquerda and direita)
```

### 🔎 Compare, diagnostique e pratique — atividade autoral

**Desafio:** formule uma regra de liberação quando `idade >= 18` **e** `cadastro_ativo` for verdadeiro; crie os quatro casos booleanos. **Diagnóstico:** qual cenário falharia se você usasse `or`?

**Registre sua resposta antes de revelar a explicação abaixo.**

### ✅ Resolução comentada — confira sua hipótese

**Alternativa correta: A.** `10 > 5` é verdadeiro e `8 < 3` é falso. Com `and`, **ambas** as condições precisam ser verdadeiras, portanto o valor da expressão é `False`, do tipo `bool`.

**Por que as demais não resolvem o problema?** Uma condição verdadeira não basta (`b`); a comparação é válida (`c`); a sintaxe Python é correta (`d`); o resultado é o booleano `False`, não o inteiro `0` (`e`).

**Critério de autoverificação:** No `or`, bastaria uma condição, abrindo acesso quando só um dos dois requisitos foi atendido.

## Questão 04 — Resto da divisão e paridade

### Enunciado

**Contextualização:** Um engenheiro de dados está implementando um algoritmo que para validar números pares em uma tabela.

**Comando:** Para verificar se um número é par através do resto da divisão por 2, o operador aritmético indicado é o

- **a)** %, retorna o resto da divisão inteira entre os operandos
- **b)** /, divisão decimal simples entre os valores
- **c)** //, extrai apenas a parte inteira do quociente final
- **d)** **, eleva o número à potência definida pelo divisor
- **e)** root(), extrai a raiz quadrada para simplificar o valor do processo

> **Primeira escolha:** marque uma alternativa provisória. Em seguida, leia o laboratório e experimente os exemplos antes de revelar o comentário de resolução.

### Antes de conferir — entenda as opções

**Operadores parecidos, tarefas distintas:** `/` calcula divisão real, `//` calcula quociente arredondado para baixo, `%` retorna o resto e `**` realiza potência. `==` é igualdade, não atribuição. Não existe operador `root()` nativo.

**💡 Dica de Ouro:** antes de escolher um operador, pergunte se você precisa do quociente, do resto ou da potência.

### Laboratório guiado

**Investigue:** compare a divisão de 11 por 2 nas quatro operações; depois descubra qual operação revela se há resto zero.

**A seguir:** execute a célula de código, observe e explique a saída; se desejar, altere os valores.

```python
for numero in (10, 11, 0, -3):
    resto = numero % 2
    print(numero, 'resto:', resto, 'par?', resto == 0)
print('11 / 2:', 11 / 2)
print('11 // 2:', 11 // 2)
print('11 % 2:', 11 % 2)
```

### 🔎 Compare, diagnostique e pratique — atividade autoral

**Desafio:** conte quantos números pares existem entre 1 e 20. Antes, compare `7 / 2`, `7 // 2` e `7 % 2`. **Diagnóstico:** qual deles informa diretamente paridade?

**Registre sua resposta antes de revelar a explicação abaixo.**

### ✅ Resolução comentada — confira sua hipótese

**Alternativa correta: A.** `%` é o operador de módulo: `numero % 2 == 0` testa se o resto ao dividir por dois é zero. A igualdade `==` é diferente da atribuição `=`.

**Por que as demais não resolvem o problema?** `/` gera o quociente; `//` produz o quociente arredondado para baixo; `**` eleva a uma potência; `root()` não é operador aritmético nativo do Python.

**Critério de autoverificação:** São 10 pares entre 1 e 20; `% 2 == 0` verifica diretamente se o resto é zero.

## Questão 05 — Classificação com `if`, `elif` e `else`

### Enunciado

**Contextualização:** Estamos criando um relatório para analisar o desempenho dos alunos; considerando a nota = 75 (essa é a nota a ser validada). O sistema verifica se é >= 90 (Excelente), >= 70 (Bom) ou inferior (Regular).

**Comando:** O resultado impresso no console pelo fluxo condicional será

- **a)** "Bom", por satisfazer a primeira condição verdadeira encontrada
- **b)** "Excelente", ignorando as condições posteriores do bloco
- **c)** "Regular", pois o sistema prioriza sempre a última cláusula de saída.
- **d)** "Bom e Regular", devido à execução em cascata de todas as opções
- **e)** "Error: Variable out of range", por não haver uma pontuação máxima definida

> **Primeira escolha:** marque uma alternativa provisória. Em seguida, leia o laboratório e experimente os exemplos antes de revelar o comentário de resolução.

### Antes de conferir — entenda as opções

**Fluxo condicional:** `if` inicia uma decisão, `elif` verifica outro caso e `else` trata os restantes. Em **uma cadeia**, só o primeiro ramo verdadeiro é executado. Dois `if` independentes não têm necessariamente o mesmo efeito.

**💡 Dica de Ouro:** para faixas com limites cumulativos, verifique da maior para a menor.

### Laboratório guiado

**Investigue:** preveja as classificações de 69, 70, 75, 89 e 90 antes de executar. Teste a ordem invertida dos limites em uma função separada.

**A seguir:** execute a célula de código, observe e explique a saída; se desejar, altere os valores.

```python
def classificar(nota):
    if nota >= 90:
        return 'Excelente'
    elif nota >= 70:
        return 'Bom'
    else:
        return 'Regular'
for nota in (65, 70, 75, 90, 95):
    print(nota, '→', classificar(nota))
```

### 🔎 Compare, diagnostique e pratique — atividade autoral

**Desafio:** implemente `classificar(69)`, `classificar(70)`, `classificar(89)` e `classificar(90)`. **Diagnóstico:** por que testar `>= 70` antes de `>= 90` daria um resultado inadequado para 95?

**Registre sua resposta antes de revelar a explicação abaixo.**

### ✅ Resolução comentada — confira sua hipótese

**Alternativa correta: A.** A nota 75 não atinge 90, mas satisfaz `>= 70`. Uma cadeia `if / elif / else` executa apenas o **primeiro ramo verdadeiro**; logo, imprime `Bom`.

**Por que as demais não resolvem o problema?** Não atinge `Excelente`; `else` não é priorizado; não executa duas alternativas da mesma cadeia; Python não impõe automaticamente um teto de notas. Se trocar a ordem para `>= 70` antes de `>= 90`, uma nota 95 será incorretamente classificada como Bom.

**Critério de autoverificação:** Limites testados do menor ao maior capturam notas altas cedo demais; 95 cairia em “Bom”.

- **a)** for, que itera diretamente sobre os elementos de uma sequência definida
- **b)** Embora possível, o while é menos legível para coleções e mais sujeito a erros de índice (loops infinitos ou erros de "out of range")

- **d)** switch, que seleciona uma coluna específica com base em um caso estático

**💡 Dica de Ouro:** quando a tarefa é percorrer elementos já disponíveis, prefira começar pelo `for`.

**Por que as demais não resolvem o problema?** O `while` poderia ser usado, mas é mais verboso nesse caso; `if` decide, não itera; `switch` não é uma estrutura de laço Python; `def` define uma função. **Complemento de apoio:** `df.columns.str.lower()` é uma alternativa para DataFrames Pandas.

## Questão 08 — Comparações com `!=`

### Enunciado

**Contextualização:** Um analista precisa gerar um relatório de marketing que exclua todos os clientes que possuem 18 anos, focando apenas nos demais.

**Comando:** Para filtrar os dados onde a idade do cliente não seja igual a 18, o operador correto é o

- **a)** !=, que valida se os valores comparados são diferentes entre si
- **b)** =, utilizado para a atribuição de novos valores às variáveis
- **c)** ==, que verifica a igualdade estrita entre dois elementos
- **d)** <>, que é uma sintaxe legada não suportada nas versões atuais
- **e)** not, que inverte o sinal matemático da variável idade para negativo

> **Primeira escolha:** marque uma alternativa provisória. Em seguida, leia o laboratório e experimente os exemplos antes de revelar o comentário de resolução.

### Antes de conferir — entenda as opções

**Comparações:** `!=` testa *diferente*, `==` testa *igual*, `=` atribui valor. `not` inverte uma condição booleana, mas não troca o sinal aritmético do número. A forma legada `<>` não é aceita em Python 3.

**💡 Dica de Ouro:** “diferente de 18” não significa “maior de 18”: são critérios de negócio distintos.

### Laboratório guiado

**Investigue:** para a mesma lista de idades, compare a filtragem por desigualdade e por igualdade. O resultado pode incluir menores de idade?

**A seguir:** execute a célula de código, observe e explique a saída; se desejar, altere os valores.

```python
idades = [15, 18, 21, 25, 18, 30]
selecionadas = []
for idade in idades:
    if idade != 18:
        selecionadas.append(idade)
print(selecionadas)
print('18 != 18:', 18 != 18)
print('21 != 18:', 21 != 18)
```

### 🔎 Compare, diagnostique e pratique — atividade autoral

**Desafio:** filtre `[17, 18, 19, 18, 20]` primeiro usando `!= 18`, depois usando `>= 18`. **Diagnóstico:** quais clientes entram no segundo resultado, mas saem do primeiro?

**Registre sua resposta antes de revelar a explicação abaixo.**

### ✅ Resolução comentada — confira sua hipótese

**Alternativa correta: A.** A comparação `idade != 18` resulta em verdadeiro para qualquer valor diferente de 18. Podemos combiná-la com `if` em um `for` ou com filtragem booleana em Pandas.

- **c)** xand, que funciona como uma conjunção exclusiva para bits

## Questão 10 — Iterações com `range()`

### Enunciado

**Contextualização:** Para treinar um modelo de IA, o desenvolvedor precisa que o ciclo de repetição execute exatamente 100 épocas de forma otimizada.

**Comando:** A função que gera uma sequência numérica eficiente para controlar o número de execuções é a

- **a)** range(), que gera os números sob demanda sem sobrecarregar a memória
- **b)** list(), que cria uma lista física na memória com todos os números
- **c)** sequence(), que ordena os dados de forma crescente em um array
- **d)** count(), que pertence à biblioteca de strings para contar caracteres
- **e)** repeat(), que reinicia a execução do código do início do arquivo

> **Primeira escolha:** marque uma alternativa provisória. Em seguida, leia o laboratório e experimente os exemplos antes de revelar o comentário de resolução.

### Antes de conferir — entenda as opções

**Iterações contadas:** `range(100)` descreve 100 inteiros de 0 a 99 sem materializar previamente uma lista de cem inteiros. `list(range(100))` cria essa lista. `sequence()` não é uma função nativa adequada ao caso; `count()` e `repeat()` existem em outros contextos, mas não reiniciam arquivos nem substituem o uso simples da faixa no laço.

**💡 Dica de Ouro:** o limite superior de `range` não entra na sequência; verifique quantas iterações realmente ocorrem.

### Laboratório guiado

**Investigue:** conte elementos em `range(100)`, `range(1,101)` e `range(0,100,2)`. Explique a diferença entre índice e número da época exibido.

**A seguir:** execute a célula de código, observe e explique a saída; se desejar, altere os valores.

```python
epocas = range(100)
print('Quantidade:', len(epocas))
print('Primeira época:', epocas[0] + 1)
print('Última época:', epocas[-1] + 1)
for epoca in range(1, 4):  # demonstração curta; equivale às épocas 1, 2, 3
    print('Época de demonstração:', epoca)
print('A estrutura range(100) controla 100 épocas quando percorrida integralmente.')
```

### 🔎 Compare, diagnostique e pratique — atividade autoral

**Desafio:** conte execuções de `range(100)`, `range(1,101)` e `range(0,100,2)` e registre primeiro/último elemento. **Diagnóstico:** a função `range()` executa o treino de IA por si só?

**Registre sua resposta antes de revelar a explicação abaixo.**

### ✅ Resolução comentada — confira sua hipótese

**Alternativa correta: A.** `range(100)` representa os inteiros de 0 a 99, perfazendo exatamente cem iterações, sem criar previamente uma lista com cem objetos inteiros. Para mostrar épocas de 1 a 100, utilize `epoca + 1` ou `range(1,101)`.

**Por que as demais não resolvem o problema?** `list()` materializa um iterável; `sequence()` não é função nativa com essa finalidade; `count()` é nome usado em diferentes métodos ou bibliotecas, não o controlador de épocas solicitado; `repeat()` não reinicia um arquivo Python.

**Critério de autoverificação:** São 100, 100 e 50 iterações. `range()` fornece valores para o `for`; não executa treinamento por conta própria.
