## Questão 07 — Tupla e imutabilidade

### Enunciado

**Contextualização:** Um analista de rotas mapeou as coordenadas geográficas centrais (Latitude e Longitude) da sede da empresa. Esses dois valores numéricos devem permanecer estritamente constantes e protegidos contra alterações acidentais ao longo de toda a execução do programa de logística.

**Comando:** Para garantir sistemicamente que essas coordenadas não possam ser modificadas pelo código após a sua declaração, os dados devem ser armazenados em uma

- **a)** Branch separada no GitHub dedicada unicamente à leitura das coordenadas
- **b)** Tupla, que atua como uma coleção de dados nativamente imutável no Python
- **c)** Variável de nome dinâmico declarada com letras maiúsculas
- **d)** Estrutura de dicionário, empregando números inteiros como chaves de acesso
- **e)** Lista, assegurada por uma estrutura if de proteção na raiz do script

**Antes de continuar:** escolha sua alternativa inicial e anote o motivo.

### Antes de conferir — entenda as opções

Tuplas são coleções imutáveis: seus itens não podem ser substituídos. Isso não torna o nome da variável uma constante absoluta: ele ainda pode ser reatribuído.

**💡 Dica de Ouro:** Uma tupla contendo objetos mutáveis não torna os objetos internos imutáveis. Para pares de números, os itens são imutáveis.

**Laboratório a seguir:** observe o comportamento e altere os valores; para operações Git, a simulação não altera seu computador.

```python
coordenadas = (-23.5505, -46.6333)
latitude, longitude = coordenadas
print('Latitude:', latitude, 'Longitude:', longitude)
try:
    coordenadas[0] = -22.9
except TypeError as erro:
    print('Alteração impedida:', type(erro).__name__)
```

**Desafio:** Teste reatribuir `coordenadas = (0, 0)`. Por que a atribuição funciona, mas `coordenadas[0] = 0` falha?

**Registre sua resposta antes de revelar a explicação abaixo.**

### ✅ Resolução comentada — confira sua hipótese

**Alternativa correta: B.** B oferece imutabilidade dos elementos da tupla. Branch não protege dados em tempo de execução; maiúsculas são convenção; dicionários e listas são mutáveis; if não torna a lista imutável.

## Questão 10 — GitFlow e branches feature

### Enunciado

**Contextualização:** Uma equipe de especialistas em dados aderiu ao padrão GitFlow para organizar seus lançamentos de software. Um analista recebe a tarefa de começar o desenvolvimento de uma nova tela de login (uma nova funcionalidade) que levará algumas semanas para ficar pronta e ser testada.

**Comando:** Seguindo a nomenclatura e o fluxo de trabalho obrigatórios do GitFlow, este trabalho deve ser iniciado através da criação de uma ramificação (branch) categorizada como

- **a)** release, originada a partir de relatórios gerenciais prévios da operação
- **b)** bugfix, puxada diretamente das pastas de backup do explorador de arquivos local
- **c)** hotfix, criada a partir da master para refletir o esforço do programador
- **d)** main, substituindo o histórico de produção diretamente com o novo código fonte
- **e)** feature, originada a partir da ramificação central de desenvolvimento develop ou main

**Antes de continuar:** escolha sua alternativa inicial e anote o motivo.

### Antes de conferir — entenda as opções

No GitFlow clássico, trabalho em nova funcionalidade usa branch `feature`, tipicamente criada de `develop`. `release` prepara lançamentos; `hotfix` corrige produção urgentemente.

**💡 Dica de Ouro:** A opção E é a melhor pelo tipo `feature`, mas a menção a `main` não é a origem canônica do GitFlow clássico: esta costuma ser `develop`.

**Laboratório a seguir:** observe o comportamento e altere os valores; para operações Git, a simulação não altera seu computador.

```python
tipos = {'feature/tela-login': 'nova funcionalidade', 'release/1.2': 'preparação de versão', 'hotfix/corrige-login': 'correção urgente de produção'}
for ramo, finalidade in tipos.items():
    print(ramo, '→', finalidade)
```

**Desafio:** Classifique três trabalhos: criar tela de login, preparar versão 2.0 e corrigir falha crítica em produção. Explique quando feature, release ou hotfix se aplica.
