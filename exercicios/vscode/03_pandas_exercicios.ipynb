### Exercício 2 — padronização antes da contagem

Use este mapa:

```python
mapa = {
    "Brasil": "Brasil",
    "Brazil": "Brasil",
    "BR": "Brasil"
}
```

Crie `pais_padronizado`. Depois compare `value_counts()` antes e depois.


````python
mapa = {
    "Brasil": "Brasil",
    "Brazil": "Brasil",
    "BR": "Brasil"
}

df_trabalho = df.copy()
df_trabalho["pais_padronizado"] = df_trabalho["pais"].replace(mapa)

print("Antes:")
print(df_trabalho["pais"].value_counts())

print()
print("Depois:")
print(df_trabalho["pais_padronizado"].value_counts())
````

<details><summary><b>Interpretação</b></summary>

Antes da padronização, três grafias podem ser contadas como três categorias.

Depois, elas representam uma única categoria analítica.

A transformação muda a forma como a informação será resumida.
</details>

# 🧪 CHECKPOINT 7 — Datas, durações e interpretação

### Exercício 1 — converter uma coluna com erro controlado

Converta `data_pedido` usando:

- `format="%Y-%m-%d"`
- `errors="coerce"`

Não sobrescreva a coluna original; crie `data_pedido_convertida`.


````python
df_datas = df.copy()

df_datas["data_pedido_convertida"] = pd.to_datetime(
    df_datas["data_pedido"],
    format="%Y-%m-%d",
    errors="coerce",
)

print(df_datas[["data_pedido", "data_pedido_convertida"]])
````

<details><summary><b>Interpretação</b></summary>

`"data_invalida"` deve virar `NaT`.

Isso não significa que a função “corrigiu” a data. Significa que ela marcou o valor como não convertido/ausente para que possamos investigar.
</details>

<details><summary><b>Resposta esperada</b></summary>

### Exercício 3 — extrair componentes com `.dt`

Crie:

- `ano`
- `mes`
- `dia_semana`

a partir de `data_pedido_convertida`.


````python
df_datas["ano"] = df_datas["data_pedido_convertida"].dt.year
df_datas["mes"] = df_datas["data_pedido_convertida"].dt.month
df_datas["dia_semana"] = df_datas["data_pedido_convertida"].dt.day_name()

print(df_datas[["data_pedido_convertida", "ano", "mes", "dia_semana"]])
````

### Exercício 4 — prazo observado

Converta `data_entrega` e calcule:

```python
prazo_dias = data_entrega - data_pedido
```

extraindo a quantidade de dias.


````python
df_datas["data_entrega_convertida"] = pd.to_datetime(
    df_datas["data_entrega"],
    errors="coerce",
)

df_datas["prazo_dias"] = (
    df_datas["data_entrega_convertida"]
    - df_datas["data_pedido_convertida"]
).dt.days

print(df_datas[
    ["data_pedido_convertida", "data_entrega_convertida", "prazo_dias"]
])
````

<details><summary><b>Leitura da expressão</b></summary>

1. selecione a data de entrega;
2. selecione a data do pedido;
3. subtraia;
4. obtenha uma duração;
5. `.dt.days` extrai os dias.

Se uma das datas estiver ausente, o prazo também fica ausente.
</details>
