### Exercício 4 — lista de dicionários, acumulador e regra

Com a lista abaixo:

- calcule o valor potencial do estoque (`preco * estoque`);
- liste apenas produtos com estoque **menor que 5**;
- mantenha o total fora do `for`.


````python
produtos = [
    {"nome": "Mouse", "preco": 89.90, "estoque": 14},
    {"nome": "Teclado", "preco": 249.90, "estoque": 7},
    {"nome": "Webcam", "preco": 179.90, "estoque": 3},
]

# Sua tentativa
````

<details><summary><b>Exemplo de solução</b></summary>

```python

for item in produtos:
    valor_estoque += item["preco"] * item["estoque"]

    if item["estoque"] < 5:
        estoque_baixo.append(item["nome"])

```

O acumulador começa antes do laço; o `if` aplica a regra em cada registro.

### Exercício 5 — mini análise sintética

**MATERIAL DIDÁTICO FICTÍCIO/SINTÉTICO.** Não é a atividade formal da professora.

Em uma lista de alunos, calcule:

- média das notas;
- quantidade de aprovados, considerando `nota >= 7` **e** `presenca >= 75`;
- nomes com presença abaixo de 75.

Use um acumulador, um contador e uma lista de resultados.


````python
alunos = [
    {"nome": "Lia", "nota": 8.0, "presenca": 90},
    {"nome": "Caio", "nota": 6.5, "presenca": 88},
    {"nome": "Nina", "nota": 7.8, "presenca": 70},
    {"nome": "Rui", "nota": 9.0, "presenca": 95},
]

# Sua tentativa
````

<details><summary><b>Exemplo de solução</b></summary>

```python
soma_notas = 0
aprovados = 0
baixa_presenca = []

for aluno in alunos:
    aprovado = aluno["nota"] >= 7 and aluno["presenca"] >= 75
    soma_notas += aluno["nota"]

    if aprovado:
        aprovados += 1

    if aluno["presenca"] < 75:
        baixa_presenca.append(aluno["nome"])

media = soma_notas / len(alunos)

print("Aprovados:", aprovados)
print(f"Média: {media:.2f}")
print("Baixa presença:", baixa_presenca)
```

</details>
