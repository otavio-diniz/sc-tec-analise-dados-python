````python
# SUA TENTATIVA — previsão de tipos e conversão.
amostras = [None, "", "0", 0, " 5 "]
# Escreva sua hipótese em uma célula Markdown antes de descomentar:
# for item in amostras: print(type(item).__name__, repr(item))
````

### ✅ Solução comentada e diagnóstico

`None` é `NoneType`; `""`, `"0"` e `" 5 "` são `str`; `0` é `int`. `float("0")` e `float(" 5 ")` convertem, mas `float("")` gera `ValueError` e `float(None)` gera `TypeError`. Tratamento seguro depende do contexto: vazio não deve ser substituído por zero sem regra explícita.

````python
# VALIDAÇÃO DIDÁTICA — solução após a tentativa
for item in [None, "", "0", 0, " 5 "]:
    try:
        print(repr(item), "→", float(item))
    except (ValueError, TypeError) as exc:
        print(repr(item), "→", type(exc).__name__, "investigar/registrar")
````

**Recuperação ativa:** quando faz sentido usar `repr`? `"0"` e `0` são intercambiáveis? Uma conversão que falha significa necessariamente um dado incorreto? **COMPREENDER / EXECUTAR / DIAGNOSTICAR**: registrar previsão, comparar execução e explicar cada diferença.
