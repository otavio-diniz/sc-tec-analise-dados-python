# 🧪 CHECKPOINT 4 — Cálculo e apresentação

### Exercício 1 — `/`, `//` e `%`

Com `10` e `3`, calcule:
1. divisão normal;
2. divisão inteira;
3. resto.

Depois explique cada resultado.


````python
a = 10
b = 3

# Sua tentativa
````

<details><summary><b>Exemplo de solução</b></summary>

```python

print(a // b)  # 3
print(a % b)   # 1
```

`/` responde o quociente; `//` quantos grupos inteiros cabem; `%` quanto sobra.

</details>

### Exercício 4 — `round()` **(complemento de apoio)** × `:.2f`

````python
valor = 4500.5
print(round(valor, 2))
print(f"{valor:.2f}")
````
