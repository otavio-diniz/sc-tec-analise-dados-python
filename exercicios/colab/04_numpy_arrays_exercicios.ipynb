## 🧪 CHECKPOINT 3 — Matrizes e eixos [AULA 16]

**N8.** Matriz fictícia de quatro pessoas por três provas: estime os tamanhos de `mean(axis=0)` e `mean(axis=1)` antes de executar.

**N9.** Extraia a última prova de todas as pessoas; encontre a maior nota por pessoa.

**N10.** Explique por que usar o eixo errado pode devolver números tecnicamente válidos e conclusões incorretas.

```python
# TENTATIVA N8–N10
notas_matriz = np.array([[8.,7.,9.],[4.,6.,5.],[7.,8.,7.],[9.,9.,8.]])
# Faça sua previsão sobre axis=0 e axis=1; confira só depois.
```

### ✅ Solução comentada CP3

Há três colunas: `mean(axis=0)` devolve três médias — uma por prova. Há quatro linhas: `mean(axis=1)` devolve quatro médias — uma por pessoa. A última coluna usa `[:, -1]`; a maior nota por pessoa usa `max(axis=1)`.

```python
print("por prova:", notas_matriz.mean(axis=0))
print("por pessoa:", notas_matriz.mean(axis=1))
print("última prova:", notas_matriz[:, -1])
print("máxima pessoa:", notas_matriz.max(axis=1))
assert len(notas_matriz.mean(axis=0)) == 3
assert len(notas_matriz.mean(axis=1)) == 4
```

## 🧪 CHECKPOINT 4 — Broadcasting e interpretação [AULA 16]

**N11.** Matriz de vendas `(2,3)` e vetor de fatores `(3,)`: preveja quais produtos receberão quais fatores. Altere a ordem dos fatores e explique por que o resultado pode permanecer executável e estar errado.

**N12.** Crie bônus diferentes para os dois dias: que forma o vetor deve adquirir para ser aplicado por linha?

**N13 [COMPLEMENTO DE APOIO].** Preveja o erro ao combinar `(2,3)` e `(2,)`, e apresente o ajuste dimensional correto para somar por linha.

```python
# TENTATIVA N11–N13
vendas_dias = np.array([[10.,20.,5.],[8.,7.,12.]])
fatores = np.array([1.1, .9, 1.2])
# Faça a previsão do shape de cada saída.
# Experimente trocar a posição dos fatores APÓS registrar a hipótese.
```

### ✅ Solução comentada CP4

O NumPy compara dimensões da direita para a esquerda. `(2,3) * (3,)` é válido por coluna; para dois bônus por dia, `(2,)[:,None]` vira `(2,1)`. Elementos alinhados por **posição**, não rótulo: o erro semântico exige verificação humana da ordem dos produtos.

```python
print("fatores por produto:", vendas_dias * fatores)
bonus_por_dia = np.array([2., 3.])[:,None]
print("bônus por dia:", vendas_dias + bonus_por_dia)
assert bonus_por_dia.shape == (2,1)
```

## 🧪 CHECKPOINT 5 — NumPy com Pandas e regras vetorizadas [AULA 16]

**N14.** Selecione três colunas numéricas e calcule média por pessoa com `axis=1`.

**N15.** Use `np.where` para classificar as médias por regra **didática e fictícia**: média ≥7, `Aprovado`; 5 a menos de 7, `Revisão`; abaixo de 5, `Reprovado`. Isso **não** representa critério oficial do curso SC TEC.

**N16.** Conte as três categorias, inclusive aquela que não ocorrer, sem inventar estudantes.

**N17.** Qual risco existe ao ordenar o DataFrame após calcular médias como array, mas antes de reinseri-las? Explique como evitá-lo.

```python
# TENTATIVA N14–N17
import pandas as pd
amostra_sint = pd.DataFrame({
    "pessoa": ["P1", "P2", "P3"],
    "nota_1": [8., 5., 9.], "nota_2": [7., 6., 9.], "nota_3": [9., 6., 8.]})
# Faça sua tentativa mantendo a ordem de linhas.
```

### ✅ Solução comentada CP5

O array de notas tem três linhas × três colunas. Uma média por pessoa requer `axis=1`. A ordem das linhas deve ser preservada ao reatribuir resultados. `np.where` produz rótulos segundo a **regra escolhida**. `value_counts().reindex(...,fill_value=0)` mostra categorias sem ocorrência, mas a contagem total deve continuar 3.

```python
colunas_nota = ["nota_1", "nota_2", "nota_3"]
valores = amostra_sint[colunas_nota].to_numpy(dtype=float)
medias = valores.mean(axis=1)
resultado = amostra_sint.copy()
resultado["media"] = medias
resultado["situacao_exemplo"] = np.where(medias >= 7, "Aprovado",
                                      np.where(medias >= 5, "Revisão", "Reprovado"))
contagens = resultado["situacao_exemplo"].value_counts().reindex(
    ["Aprovado","Revisão","Reprovado"], fill_value=0)
print(resultado.to_string(index=False))
print(contagens)
assert contagens.sum() == len(resultado)
```
